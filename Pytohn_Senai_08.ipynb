```python
def verifica_numero(numero):
    if numero > 0:
        return 'P'
    else:
        return 'N'


# Programa principal
n = int(input("Digite um número: "))
print(f"O resultado é: {verifica_numero(n)}")
```

```python
def somaImposto(taxaImposto, custo):
    # taxaImposto vem em porcentagem (ex: 10 = 10%)
    custo_com_imposto = custo + (custo * taxaImposto / 100)
    return custo_com_imposto


# Programa principal
taxa = float(input("Digite a taxa de imposto (%): "))
custo = float(input("Digite o custo do produto (R$): "))

novo_valor = somaImposto(taxa, custo)
print(f"O valor final com imposto é: R$ {novo_valor:.2f}")
```

```python
def valorPagamento(valor, dias):
    if dias == 0:
        return valor
    else:
        multa = valor * 0.03                 # 3% de multa
        juros = valor * 0.001 * dias         # 0,1% ao dia
        return valor + multa + juros


# Programa principal
total_pago = 0
quantidade_prestacoes = 0

while True:
    valor = float(input("Digite o valor da prestação (0 para encerrar): "))
    if valor == 0:
        break

    dias = int(input("Digite o número de dias em atraso: "))
    valor_final = valorPagamento(valor, dias)

    print(f"Valor a ser pago: R$ {valor_final:.2f}\n")

    quantidade_prestacoes += 1
    total_pago += valor_final

# Relatório final
print("===== RELATÓRIO DO DIA =====")
print(f"Quantidade de prestações pagas: {quantidade_prestacoes}")
print(f"Valor total pago: R$ {total_pago:.2f}")
```
