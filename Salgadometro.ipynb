```python
import tkinter as tk
from tkinter import messagebox
import random

# Dados dos produtos
produtos = {
    "001": {"nome": "Esfiha Carne", "valor": 1.6},
    "002": {"nome": "Esfiha Calabresa", "valor": 1.6},
    "003": {"nome": "Mini pizza Calabresa", "valor": 1.7},
    "004": {"nome": "Parmesão com Tomate Seco", "valor": 1.6},
    "005": {"nome": "Quiche de Queijo com Bacon", "valor": 1.2}
}

# Lista para armazenar o pedido do usuário
pedido = []

# Função para adicionar o produto selecionado ao pedido
def adicionar_ao_pedido():
    indice_selecionado = listbox_produtos.curselection()
    if not indice_selecionado:
        messagebox.showwarning("Aviso", "Selecione um produto para adicionar ao pedido.")
        return

    codigo_produto = listbox_produtos.get(indice_selecionado).split(" ")[0]
    produto = produtos[codigo_produto]

    try:
        quantidade = int(entry_quantidade.get())
        if quantidade <= 0:
            raise ValueError("Quantidade deve ser maior que zero.")

        item_pedido = {"nome": produto["nome"], "valor": produto["valor"], "quantidade": quantidade}
        pedido.append(item_pedido)
        atualizar_pedido()

    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira uma quantidade válida.")

def atualizar_pedido():
    listbox_pedido.delete(0, tk.END)
    total_valor = 0
    total_salgados = sum(item["quantidade"] for item in pedido)

    for item in pedido:
        total_item = item["valor"] * item["quantidade"]
        listbox_pedido.insert(tk.END, f"{item['nome']} - {item['quantidade']}x (R$ {total_item:.2f})")
        total_valor += total_item

    try:
        valor_entrega = float(entry_valor_entrega.get())
    except ValueError:
        valor_entrega = 0

    total_valor_com_entrega = total_valor + valor_entrega

    label_total_salgados_var.set(f"Total de salgados no pedido: {total_salgados}")
    label_total_valor_var.set(f"Valor total do pedido: R$ {total_valor_com_entrega:.2f}")

    # Atualiza a comparação entre a média e o total de salgados
    comparar_media_salgados(total_salgados)

def comparar_media_salgados(total_salgados):
    try:
        quantidade_pessoas = int(entry_quantidade_pessoas.get())
        media_salgados_por_pessoa = float(entry_media_salgados.get())

        media_total = quantidade_pessoas * media_salgados_por_pessoa

        if total_salgados < media_total:
            resultado = "Menos do que a média recomendada."
        elif total_salgados > media_total:
            resultado = "Mais do que a média recomendada."
        else:
            resultado = "Igual à média recomendada."

        label_comparacao_var.set(f"Comparação: {resultado} (Média: {media_total}, Total: {total_salgados})")

    except ValueError:
        label_comparacao_var.set("Preencha a quantidade de pessoas e média para comparar.")

def calcular_salgados():
    try:
        quantidade_pessoas = int(entry_quantidade_pessoas.get())
        media_salgados_por_pessoa = float(entry_media_salgados.get())

        total_salgados = quantidade_pessoas * media_salgados_por_pessoa
        label_total_salgados_calculado_var.set(f"Total de salgados necessário: {total_salgados}")

    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira valores numéricos válidos.")

def atualizar_lista_produtos():
    listbox_produtos.delete(0, tk.END)
    for codigo, detalhes in produtos.items():
        produto_info = f"{codigo} - {detalhes['nome']} (R$ {detalhes['valor']:.2f})"
        listbox_produtos.insert(tk.END, produto_info)

# Função para calcular a quantidade máxima aleatória e o valor máximo
def calcular_maximo_aleatorio():
    try:
        # Limpa a lista de pedidos antes de calcular
        pedido.clear()

        # Obtém a quantidade máxima e o valor máximo do usuário
        quantidade_maxima = int(entry_quantidade_maxima.get())
        valor_maximo = float(entry_valor_maximo.get())

        total_valor = 0
        total_quantidade = 0

        # Gera produtos aleatórios até atingir os limites
        while total_quantidade < quantidade_maxima and total_valor < valor_maximo:
            codigo_produto = random.choice(list(produtos.keys()))
            produto = produtos[codigo_produto]
            quantidade = random.randint(1, 10)  # quantidade aleatória de cada produto

            # Verifica se a nova quantidade e valor não ultrapassam os limites
            if (total_quantidade + quantidade <= quantidade_maxima) and (total_valor + produto["valor"] * quantidade <= valor_maximo):
                item_pedido = {"nome": produto["nome"], "valor": produto["valor"], "quantidade": quantidade}
                pedido.append(item_pedido)
                total_quantidade += quantidade
                total_valor += produto["valor"] * quantidade
            else:
                # Se não for possível adicionar mais itens, interrompe o loop
                break

        atualizar_pedido()  # Atualiza a lista de pedidos
    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira valores válidos.")

# Cria a janela principal
janela = tk.Tk()
janela.title("Calculadora de Salgados")

# Frame principal
frame_principal = tk.Frame(janela)
frame_principal.pack(padx=10, pady=10, fill=tk.BOTH, expand=True)

# Canvas para permitir a rolagem
canvas = tk.Canvas(frame_principal)
canvas.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)

# Scrollbar
scrollbar = tk.Scrollbar(frame_principal, orient="vertical", command=canvas.yview)
scrollbar.pack(side=tk.RIGHT, fill=tk.Y)

# Configura o canvas
canvas.configure(yscrollcommand=scrollbar.set)
canvas.bind('<Configure>', lambda e: canvas.configure(scrollregion=canvas.bbox("all")))

# Frame interno para conter os widgets
frame_interno = tk.Frame(canvas)
canvas.create_window((0, 0), window=frame_interno, anchor="nw")

# Adicionando widgets ao frame interno
# Campo para a quantidade de pessoas
tk.Label(frame_interno, text="Qtd. de Pessoas:").grid(row=0, column=0)
entry_quantidade_pessoas = tk.Entry(frame_interno)
entry_quantidade_pessoas.grid(row=0, column=1)

# Campo para a média de salgados por pessoa
tk.Label(frame_interno, text="Média Salgados/Pessoa:").grid(row=1, column=0)
entry_media_salgados = tk.Entry(frame_interno)
entry_media_salgados.insert(0, "17")
entry_media_salgados.grid(row=1, column=1)

# Botão para calcular total de salgados
tk.Button(frame_interno, text="Calcular Salgados", command=calcular_salgados).grid(row=2, column=0, columnspan=2)

# Label para exibir total de salgados calculados
label_total_salgados_calculado_var = tk.StringVar()
label_total_salgados_calculado_var.set("Total de salgados necessário: 0")
tk.Label(frame_interno, textvariable=label_total_salgados_calculado_var).grid(row=3, column=0, columnspan=2)

# Lista de produtos com scrollbar
tk.Label(frame_interno, text="Produtos Disponíveis:").grid(row=4, column=0, columnspan=2)

frame_produtos = tk.Frame(frame_interno)
frame_produtos.grid(row=5, column=0, columnspan=2)

listbox_produtos = tk.Listbox(frame_produtos, width=50, height=5)
listbox_produtos.pack(side=tk.LEFT, fill=tk.BOTH)

scrollbar_produtos = tk.Scrollbar(frame_produtos)
scrollbar_produtos.pack(side=tk.RIGHT, fill=tk.Y)

listbox_produtos.config(yscrollcommand=scrollbar_produtos.set)
scrollbar_produtos.config(command=listbox_produtos.yview)

atualizar_lista_produtos()  # Atualiza a lista de produtos no início

# Campo para a quantidade a ser adicionada
tk.Label(frame_interno, text="Quantidade a Adicionar:").grid(row=6, column=0)
entry_quantidade = tk.Entry(frame_interno)
entry_quantidade.grid(row=6, column=1)

# Botão para adicionar ao pedido
tk.Button(frame_interno, text="Adicionar ao Pedido", command=adicionar_ao_pedido).grid(row=7, column=0, columnspan=2)

# Lista do pedido
tk.Label(frame_interno, text="Pedido:").grid(row=8, column=0, columnspan=2)

listbox_pedido = tk.Listbox(frame_interno, width=50, height=5)
listbox_pedido.grid(row=9, column=0, columnspan=2)

# Campo para valor de entrega
tk.Label(frame_interno, text="Valor da Entrega:").grid(row=10, column=0)
entry_valor_entrega = tk.Entry(frame_interno)
entry_valor_entrega.grid(row=10, column=1)

# Labels para exibir totais
label_total_salgados_var = tk.StringVar()
label_total_salgados_var.set("Total de salgados no pedido: 0")
tk.Label(frame_interno, textvariable=label_total_salgados_var).grid(row=11, column=0, columnspan=2)

label_total_valor_var = tk.StringVar()
label_total_valor_var.set("Valor total do pedido: R$ 0.00")
tk.Label(frame_interno, textvariable=label_total_valor_var).grid(row=12, column=0, columnspan=2)

# Campos para calcular aleatório
tk.Label(frame_interno, text="Qtd. Máxima de Produtos:").grid(row=13, column=0)
entry_quantidade_maxima = tk.Entry(frame_interno)
entry_quantidade_maxima.grid(row=13, column=1)

tk.Label(frame_interno, text="Valor Máximo Total:").grid(row=14, column=0)
entry_valor_maximo = tk.Entry(frame_interno)
entry_valor_maximo.grid(row=14, column=1)

# Botão para calcular aleatório
tk.Button(frame_interno, text="Calcular Aleatório", command=calcular_maximo_aleatorio).grid(row=15, column=0, columnspan=2)

# Campo para adicionar novos produtos
tk.Label(frame_interno, text="Nome do Produto:").grid(row=16, column=0)
entry_nome_produto = tk.Entry(frame_interno)
entry_nome_produto.grid(row=16, column=1)

tk.Label(frame_interno, text="Valor do Produto:").grid(row=17, column=0)
entry_valor_produto = tk.Entry(frame_interno)
entry_valor_produto.grid(row=17, column=1)

# Botão para adicionar novo produto
tk.Button(frame_interno, text="Adicionar Produto", command=lambda: adicionar_produto()).grid(row=18, column=0, columnspan=2)

# Label para exibir comparação
label_comparacao_var = tk.StringVar()
label_comparacao_var.set("Comparação: Preencha a quantidade de pessoas e média para comparar.")
tk.Label(frame_interno, textvariable=label_comparacao_var).grid(row=19, column=0, columnspan=2)

def adicionar_produto():
    nome = entry_nome_produto.get()
    valor = entry_valor_produto.get()

    if not nome or not valor:
        messagebox.showwarning("Aviso", "Por favor, preencha todos os campos.")
        return

    try:
        valor_float = float(valor)
        codigo = f"{len(produtos) + 1:03}"
        produtos[codigo] = {"nome": nome, "valor": valor_float}
        atualizar_lista_produtos()
        entry_nome_produto.delete(0, tk.END)
        entry_valor_produto.delete(0, tk.END)
    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira um valor válido para o produto.")

# Inicia o loop da interface
janela.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox
import random

# Dados dos produtos
produtos = {
    "001": {"nome": "Esfiha Carne", "valor": 1.6},
    "002": {"nome": "Esfiha Calabresa", "valor": 1.6},
    "003": {"nome": "Mini pizza Calabresa", "valor": 1.7},
    "004": {"nome": "Parmesão com Tomate Seco", "valor": 1.6},
    "005": {"nome": "Quiche de Queijo com Bacon", "valor": 1.2}
}

# Lista para armazenar o pedido do usuário
pedido = []

# Função para adicionar o produto selecionado ao pedido
def adicionar_ao_pedido():
    indice_selecionado = listbox_produtos.curselection()
    if not indice_selecionado:
        messagebox.showwarning("Aviso", "Selecione um produto para adicionar ao pedido.")
        return

    codigo_produto = listbox_produtos.get(indice_selecionado).split(" ")[0]
    produto = produtos[codigo_produto]

    try:
        quantidade = int(entry_quantidade.get())
        if quantidade <= 0:
            raise ValueError("Quantidade deve ser maior que zero.")

        item_pedido = {"nome": produto["nome"], "valor": produto["valor"], "quantidade": quantidade}
        pedido.append(item_pedido)
        atualizar_pedido()

    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira uma quantidade válida.")

def atualizar_pedido():
    listbox_pedido.delete(0, tk.END)
    total_valor = 0
    total_salgados = sum(item["quantidade"] for item in pedido)

    for item in pedido:
        total_item = item["valor"] * item["quantidade"]
        listbox_pedido.insert(tk.END, f"{item['nome']} - {item['quantidade']}x (R$ {total_item:.2f})")
        total_valor += total_item

    try:
        valor_entrega = float(entry_valor_entrega.get())
    except ValueError:
        valor_entrega = 0

    total_valor_com_entrega = total_valor + valor_entrega

    label_total_salgados_var.set(f"Total de salgados no pedido: {total_salgados}")
    label_total_valor_var.set(f"Valor total do pedido: R$ {total_valor_com_entrega:.2f}")

    # Atualiza a comparação entre a média e o total de salgados
    comparar_media_salgados(total_salgados)
    # Atualiza o cálculo da média real de salgados por pessoa
    calcular_media_real(total_salgados)

def comparar_media_salgados(total_salgados):
    try:
        total_salgados_necessario = float(label_total_salgados_calculado_var.get().split(": ")[1])
        if total_salgados_necessario > 0:
            media_ideal = total_salgados / total_salgados_necessario

            if media_ideal < 1:
                resultado = "Menos do que a média recomendada."
                label_comparacao.config(fg="red")  # Define a cor do texto para vermelho
            elif media_ideal > 1:
                resultado = "Mais do que a média recomendada."
                label_comparacao.config(fg="black")  # Define a cor do texto para preto
            else:
                resultado = "Igual à média recomendada."
                label_comparacao.config(fg="black")  # Define a cor do texto para preto

            label_comparacao_var.set(f"Comparação: {resultado} (Média Ideal: {media_ideal:.2f}, Total: {total_salgados})")
        else:
            label_comparacao_var.set("Preencha a quantidade de pessoas e média para calcular o total de salgados necessário.")
    except ValueError:
        label_comparacao_var.set("Preencha a quantidade de pessoas e média para calcular o total de salgados necessário.")

def calcular_salgados():
    try:
        quantidade_pessoas = int(entry_quantidade_pessoas.get())
        media_salgados_por_pessoa = float(entry_media_salgados.get())

        total_salgados = quantidade_pessoas * media_salgados_por_pessoa
        label_total_salgados_calculado_var.set(f"Total de salgados necessário: {total_salgados}")

    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira valores numéricos válidos.")

def calcular_media_real(total_salgados):
    try:
        quantidade_pessoas = int(entry_quantidade_pessoas.get())
        if quantidade_pessoas > 0:
            media_real = total_salgados / quantidade_pessoas
            label_media_real_var.set(f"Média real de salgados por pessoa: {media_real:.2f}")
        else:
            label_media_real_var.set("Média real de salgados por pessoa: N/A")
    except ValueError:
        label_media_real_var.set("Média real de salgados por pessoa: N/A")

def atualizar_lista_produtos():
    listbox_produtos.delete(0, tk.END)
    for codigo, detalhes in produtos.items():
        produto_info = f"{codigo} - {detalhes['nome']} (R$ {detalhes['valor']:.2f})"
        listbox_produtos.insert(tk.END, produto_info)

def agrupar_pedido():
    agrupado = {}
    for item in pedido:
        if item["nome"] in agrupado:
            agrupado[item["nome"]]["quantidade"] += item["quantidade"]
        else:
            agrupado[item["nome"]] = {"valor": item["valor"], "quantidade": item["quantidade"]}
    
    # Limpa o pedido atual e adiciona os itens agrupados
    pedido.clear()
    for nome, detalhes in agrupado.items():
        pedido.append({"nome": nome, "valor": detalhes["valor"], "quantidade": detalhes["quantidade"]})
    
    atualizar_pedido()

def calcular_maximo_aleatorio():
    try:
        # Limpa a lista de pedidos antes de calcular
        pedido.clear()

        # Obtém a quantidade máxima e o valor máximo do usuário
        quantidade_maxima = int(entry_quantidade_maxima.get())
        valor_maximo = float(entry_valor_maximo.get())

        total_valor = 0
        total_quantidade = 0

        # Gera produtos aleatórios até atingir os limites
        while total_quantidade < quantidade_maxima and total_valor < valor_maximo:
            codigo_produto = random.choice(list(produtos.keys()))
            produto = produtos[codigo_produto]
            quantidade = random.randint(1, 10)  # quantidade aleatória de cada produto

            # Verifica se a nova quantidade e valor não ultrapassam os limites
            if (total_quantidade + quantidade <= quantidade_maxima) and (total_valor + produto["valor"] * quantidade <= valor_maximo):
                item_pedido = {"nome": produto["nome"], "valor": produto["valor"], "quantidade": quantidade}
                pedido.append(item_pedido)
                total_quantidade += quantidade
                total_valor += produto["valor"] * quantidade
            else:
                # Se não for possível adicionar mais itens, interrompe o loop
                break

        atualizar_pedido()  # Atualiza a lista de pedidos
    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira valores válidos.")

# Cria a janela principal
janela = tk.Tk()
janela.title("Calculadora de Salgados")

# Frame principal
frame_principal = tk.Frame(janela)
frame_principal.pack(padx=10, pady=10, fill=tk.BOTH, expand=True)

# Canvas para permitir a rolagem
canvas = tk.Canvas(frame_principal)
canvas.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)

# Scrollbar
scrollbar = tk.Scrollbar(frame_principal, orient="vertical", command=canvas.yview)
scrollbar.pack(side=tk.RIGHT, fill=tk.Y)

# Configura o canvas
canvas.configure(yscrollcommand=scrollbar.set)
canvas.bind('<Configure>', lambda e: canvas.configure(scrollregion=canvas.bbox("all")))

# Frame interno para conter os widgets
frame_interno = tk.Frame(canvas)
canvas.create_window((0, 0), window=frame_interno, anchor="nw")

# Adicionando widgets ao frame interno
# Campo para a quantidade de pessoas
tk.Label(frame_interno, text="Qtd. de Pessoas:").grid(row=0, column=0)
entry_quantidade_pessoas = tk.Entry(frame_interno)
entry_quantidade_pessoas.grid(row=0, column=1)

# Campo para a média de salgados por pessoa
tk.Label(frame_interno, text="Média Salgados/Pessoa:").grid(row=1, column=0)
entry_media_salgados = tk.Entry(frame_interno)
entry_media_salgados.insert(0, "17")
entry_media_salgados.grid(row=1, column=1)

# Botão para calcular total de salgados
tk.Button(frame_interno, text="Calcular Salgados", command=calcular_salgados).grid(row=2, column=0, columnspan=2)

# Label para exibir total de salgados calculados
label_total_salgados_calculado_var = tk.StringVar()
label_total_salgados_calculado_var.set("Total de salgados necessário: 0")
tk.Label(frame_interno, textvariable=label_total_salgados_calculado_var).grid(row=3, column=0, columnspan=2)

# Lista de produtos com scrollbar
tk.Label(frame_interno, text="Produtos Disponíveis:").grid(row=4, column=0, columnspan=2)

frame_produtos = tk.Frame(frame_interno)
frame_produtos.grid(row=5, column=0, columnspan=2)

listbox_produtos = tk.Listbox(frame_produtos, width=50, height=5)
listbox_produtos.pack(side=tk.LEFT, fill=tk.BOTH)

scrollbar_produtos = tk.Scrollbar(frame_produtos)
scrollbar_produtos.pack(side=tk.RIGHT, fill=tk.Y)

listbox_produtos.config(yscrollcommand=scrollbar_produtos.set)
scrollbar_produtos.config(command=listbox_produtos.yview)

atualizar_lista_produtos()  # Atualiza a lista de produtos no início

# Campo para a quantidade a ser adicionada
tk.Label(frame_interno, text="Quantidade a Adicionar:").grid(row=6, column=0)
entry_quantidade = tk.Entry(frame_interno)
entry_quantidade.grid(row=6, column=1)

# Botão para adicionar ao pedido
tk.Button(frame_interno, text="Adicionar ao Pedido", command=adicionar_ao_pedido).grid(row=7, column=0, columnspan=2)

# Lista do pedido
tk.Label(frame_interno, text="Pedido:").grid(row=8, column=0, columnspan=2)

listbox_pedido = tk.Listbox(frame_interno, width=50, height=5)
listbox_pedido.grid(row=9, column=0, columnspan=2)

# Campo para valor de entrega
tk.Label(frame_interno, text="Valor da Entrega:").grid(row=10, column=0)
entry_valor_entrega = tk.Entry(frame_interno)
entry_valor_entrega.grid(row=10, column=1)

# Labels para exibir totais
label_total_salgados_var = tk.StringVar()
label_total_salgados_var.set("Total de salgados no pedido: 0")
tk.Label(frame_interno, textvariable=label_total_salgados_var).grid(row=11, column=0, columnspan=2)

label_total_valor_var = tk.StringVar()
label_total_valor_var.set("Valor total do pedido: R$ 0.00")
tk.Label(frame_interno, textvariable=label_total_valor_var).grid(row=12, column=0, columnspan=2)

# Label para exibir a média real de salgados por pessoa
label_media_real_var = tk.StringVar()
label_media_real_var.set("Média real de salgados por pessoa: N/A")
tk.Label(frame_interno, textvariable=label_media_real_var).grid(row=13, column=0, columnspan=2)

# Label para exibir a comparação de média de salgados
label_comparacao_var = tk.StringVar()
label_comparacao_var.set("Comparação: N/A")
label_comparacao = tk.Label(frame_interno, textvariable=label_comparacao_var)
label_comparacao.grid(row=14, column=0, columnspan=2)

# Botão para agrupar itens do pedido
tk.Button(frame_interno, text="Agrupar Itens do Pedido", command=agrupar_pedido).grid(row=15, column=0, columnspan=2)

# Campos para calcular aleatório
tk.Label(frame_interno, text="Qtd. Máxima de Produtos:").grid(row=16, column=0)
entry_quantidade_maxima = tk.Entry(frame_interno)
entry_quantidade_maxima.grid(row=16, column=1)

tk.Label(frame_interno, text="Valor Máximo Total:").grid(row=17, column=0)
entry_valor_maximo = tk.Entry(frame_interno)
entry_valor_maximo.grid(row=17, column=1)

# Botão para calcular aleatório
tk.Button(frame_interno, text="Calcular Aleatório", command=calcular_maximo_aleatorio).grid(row=18, column=0, columnspan=2)

# Inicia o loop da interface
janela.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox
import random

# Dados dos produtos
produtos = {
    "001": {"nome": "Esfiha Carne", "valor": 1.6},
    "002": {"nome": "Esfiha Calabresa", "valor": 1.6},
    "003": {"nome": "Mini pizza Calabresa", "valor": 1.7},
    "004": {"nome": "Parmesão com Tomate Seco", "valor": 1.6},
    "005": {"nome": "Quiche de Queijo com Bacon", "valor": 1.2}
}

# Lista para armazenar o pedido do usuário
pedido = []

# Função para adicionar um novo produto ao dicionário 'produtos'
def adicionar_produto():
    codigo = entry_codigo_produto.get().strip()
    nome = entry_nome_produto.get().strip()
    valor = entry_valor_produto.get().strip()

    if not codigo or not nome or not valor:
        messagebox.showwarning("Aviso", "Todos os campos devem ser preenchidos para adicionar um novo produto.")
        return

    try:
        valor = float(valor)
        if valor <= 0:
            raise ValueError("O valor do produto deve ser maior que zero.")
    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira um valor válido para o produto.")
        return

    if codigo in produtos:
        messagebox.showwarning("Aviso", "Código já existente. Escolha um código diferente.")
        return

    produtos[codigo] = {"nome": nome, "valor": valor}
    atualizar_lista_produtos()
    messagebox.showinfo("Sucesso", f"Produto '{nome}' adicionado com sucesso!")

# Função para adicionar o produto selecionado ao pedido
def adicionar_ao_pedido():
    indice_selecionado = listbox_produtos.curselection()
    if not indice_selecionado:
        messagebox.showwarning("Aviso", "Selecione um produto para adicionar ao pedido.")
        return

    codigo_produto = listbox_produtos.get(indice_selecionado).split(" ")[0]
    produto = produtos[codigo_produto]

    try:
        quantidade = int(entry_quantidade.get())
        if quantidade <= 0:
            raise ValueError("Quantidade deve ser maior que zero.")

        item_pedido = {"nome": produto["nome"], "valor": produto["valor"], "quantidade": quantidade}
        pedido.append(item_pedido)
        atualizar_pedido()

    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira uma quantidade válida.")

# Função para atualizar o pedido
def atualizar_pedido():
    listbox_pedido.delete(0, tk.END)
    total_valor = 0
    total_salgados = sum(item["quantidade"] for item in pedido)

    for item in pedido:
        total_item = item["valor"] * item["quantidade"]
        listbox_pedido.insert(tk.END, f"{item['nome']} - {item['quantidade']}x (R$ {total_item:.2f})")
        total_valor += total_item

    try:
        valor_entrega = float(entry_valor_entrega.get())
    except ValueError:
        valor_entrega = 0

    total_valor_com_entrega = total_valor + valor_entrega

    label_total_salgados_var.set(f"Total de salgados no pedido: {total_salgados}")
    label_total_valor_var.set(f"Valor total do pedido: R$ {total_valor_com_entrega:.2f}")

    # Atualiza a comparação entre a média e o total de salgados
    comparar_media_salgados(total_salgados)
    # Atualiza o cálculo da média real de salgados por pessoa
    calcular_media_real(total_salgados)

# Função para comparar a média de salgados
def comparar_media_salgados(total_salgados):
    try:
        total_salgados_necessario = float(label_total_salgados_calculado_var.get().split(": ")[1])
        if total_salgados_necessario > 0:
            media_ideal = total_salgados / total_salgados_necessario

            if media_ideal < 1:
                resultado = "Menos do que a média recomendada."
                label_comparacao.config(fg="red")  # Define a cor do texto para vermelho
            elif media_ideal > 1:
                resultado = "Mais do que a média recomendada."
                label_comparacao.config(fg="black")  # Define a cor do texto para preto
            else:
                resultado = "Igual à média recomendada."
                label_comparacao.config(fg="black")  # Define a cor do texto para preto

            label_comparacao_var.set(f"Comparação: {resultado} (Média Ideal: {media_ideal:.2f}, Total: {total_salgados})")
        else:
            label_comparacao_var.set("Preencha a quantidade de pessoas e média para calcular o total de salgados necessário.")
    except ValueError:
        label_comparacao_var.set("Preencha a quantidade de pessoas e média para calcular o total de salgados necessário.")

# Função para calcular o total de salgados necessários
def calcular_salgados():
    try:
        quantidade_pessoas = int(entry_quantidade_pessoas.get())
        media_salgados_por_pessoa = float(entry_media_salgados.get())

        total_salgados = quantidade_pessoas * media_salgados_por_pessoa
        label_total_salgados_calculado_var.set(f"Total de salgados necessário: {total_salgados}")

    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira valores numéricos válidos.")

# Função para calcular a média real de salgados por pessoa
def calcular_media_real(total_salgados):
    try:
        quantidade_pessoas = int(entry_quantidade_pessoas.get())
        if quantidade_pessoas > 0:
            media_real = total_salgados / quantidade_pessoas
            label_media_real_var.set(f"Média real de salgados por pessoa: {media_real:.2f}")
        else:
            label_media_real_var.set("Média real de salgados por pessoa: N/A")
    except ValueError:
        label_media_real_var.set("Média real de salgados por pessoa: N/A")

# Atualiza a lista de produtos na interface
def atualizar_lista_produtos():
    listbox_produtos.delete(0, tk.END)
    for codigo, detalhes in produtos.items():
        produto_info = f"{codigo} - {detalhes['nome']} (R$ {detalhes['valor']:.2f})"
        listbox_produtos.insert(tk.END, produto_info)

# Função para agrupar os itens do pedido
def agrupar_pedido():
    agrupado = {}
    for item in pedido:
        if item["nome"] in agrupado:
            agrupado[item["nome"]]["quantidade"] += item["quantidade"]
        else:
            agrupado[item["nome"]] = {"valor": item["valor"], "quantidade": item["quantidade"]}
    
    # Limpa o pedido atual e adiciona os itens agrupados
    pedido.clear()
    for nome, detalhes in agrupado.items():
        pedido.append({"nome": nome, "valor": detalhes["valor"], "quantidade": detalhes["quantidade"]})
    
    atualizar_pedido()

# Função para calcular um pedido aleatório respeitando limites
def calcular_maximo_aleatorio():
    try:
        # Limpa a lista de pedidos antes de calcular
        pedido.clear()

        # Obtém a quantidade máxima e o valor máximo do usuário
        quantidade_maxima = int(entry_quantidade_maxima.get())
        valor_maximo = float(entry_valor_maximo.get())

        total_valor = 0
        total_quantidade = 0

        # Gera produtos aleatórios até atingir os limites
        while total_quantidade < quantidade_maxima and total_valor < valor_maximo:
            codigo_produto = random.choice(list(produtos.keys()))
            produto = produtos[codigo_produto]
            quantidade = random.randint(1, 10)  # quantidade aleatória de cada produto

            # Verifica se a nova quantidade e valor não ultrapassam os limites
            if (total_quantidade + quantidade <= quantidade_maxima) and (total_valor + produto["valor"] * quantidade <= valor_maximo):
                item_pedido = {"nome": produto["nome"], "valor": produto["valor"], "quantidade": quantidade}
                pedido.append(item_pedido)
                total_quantidade += quantidade
                total_valor += produto["valor"] * quantidade
            else:
                # Se não for possível adicionar mais itens, interrompe o loop
                break

        atualizar_pedido()  # Atualiza a lista de pedidos
    except ValueError:
        messagebox.showerror("Erro", "Por favor, insira valores válidos.")

# Cria a janela principal
janela = tk.Tk()
janela.title("Calculadora de Salgados")

# Frame principal
frame_principal = tk.Frame(janela)
frame_principal.pack(padx=10, pady=10, fill=tk.BOTH, expand=True)

# Canvas para permitir a rolagem
canvas = tk.Canvas(frame_principal)
canvas.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)

# Scrollbar
scrollbar = tk.Scrollbar(frame_principal, orient="vertical", command=canvas.yview)
scrollbar.pack(side=tk.RIGHT, fill=tk.Y)

# Configura o canvas
canvas.configure(yscrollcommand=scrollbar.set)
canvas.bind('<Configure>', lambda e: canvas.configure(scrollregion=canvas.bbox("all")))

# Frame interno para conter os widgets
frame_interno = tk.Frame(canvas)
canvas.create_window((0, 0), window=frame_interno, anchor="nw")

# Adicionando widgets ao frame interno
# Campo para a quantidade de pessoas
tk.Label(frame_interno, text="Qtd. de Pessoas:").grid(row=0, column=0)
entry_quantidade_pessoas = tk.Entry(frame_interno)
entry_quantidade_pessoas.grid(row=0, column=1)

# Campo para a média de salgados por pessoa
tk.Label(frame_interno, text="Média Salgados/Pessoa:").grid(row=1, column=0)
entry_media_salgados = tk.Entry(frame_interno)
entry_media_salgados.insert(0, "17")
entry_media_salgados.grid(row=1, column=1)

# Botão para calcular total de salgados
tk.Button(frame_interno, text="Calcular Salgados", command=calcular_salgados).grid(row=2, column=0, columnspan=2)

# Label para exibir total de salgados calculados
label_total_salgados_calculado_var = tk.StringVar()
label_total_salgados_calculado_var.set("Total de salgados necessário: 0")
tk.Label(frame_interno, textvariable=label_total_salgados_calculado_var).grid(row=3, column=0, columnspan=2)

# Lista de produtos com scrollbar
tk.Label(frame_interno, text="Produtos Disponíveis:").grid(row=4, column=0, columnspan=2)

frame_produtos = tk.Frame(frame_interno)
frame_produtos.grid(row=5, column=0, columnspan=2)

listbox_produtos = tk.Listbox(frame_produtos, width=50, height=5)
listbox_produtos.pack(side=tk.LEFT, fill=tk.BOTH)

scrollbar_produtos = tk.Scrollbar(frame_produtos)
scrollbar_produtos.pack(side=tk.RIGHT, fill=tk.Y)

listbox_produtos.config(yscrollcommand=scrollbar_produtos.set)
scrollbar_produtos.config(command=listbox_produtos.yview)

atualizar_lista_produtos()  # Atualiza a lista de produtos no início

# Campo para a quantidade a ser adicionada
tk.Label(frame_interno, text="Quantidade a Adicionar:").grid(row=6, column=0)
entry_quantidade = tk.Entry(frame_interno)
entry_quantidade.grid(row=6, column=1)

# Botão para adicionar ao pedido
tk.Button(frame_interno, text="Adicionar ao Pedido", command=adicionar_ao_pedido).grid(row=7, column=0, columnspan=2)

# Lista do pedido
tk.Label(frame_interno, text="Pedido:").grid(row=8, column=0, columnspan=2)

listbox_pedido = tk.Listbox(frame_interno, width=50, height=5)
listbox_pedido.grid(row=9, column=0, columnspan=2)

# Campo para valor de entrega
tk.Label(frame_interno, text="Valor da Entrega:").grid(row=10, column=0)
entry_valor_entrega = tk.Entry(frame_interno)
entry_valor_entrega.grid(row=10, column=1)

# Labels para exibir totais
label_total_salgados_var = tk.StringVar()
label_total_salgados_var.set("Total de salgados no pedido: 0")
tk.Label(frame_interno, textvariable=label_total_salgados_var).grid(row=11, column=0, columnspan=2)

label_total_valor_var = tk.StringVar()
label_total_valor_var.set("Valor total do pedido: R$ 0.00")
tk.Label(frame_interno, textvariable=label_total_valor_var).grid(row=12, column=0, columnspan=2)

# Label para exibir a média real de salgados por pessoa
label_media_real_var = tk.StringVar()
label_media_real_var.set("Média real de salgados por pessoa: N/A")
tk.Label(frame_interno, textvariable=label_media_real_var).grid(row=13, column=0, columnspan=2)

# Label para exibir a comparação de média de salgados
label_comparacao_var = tk.StringVar()
label_comparacao_var.set("Comparação: N/A")
label_comparacao = tk.Label(frame_interno, textvariable=label_comparacao_var)
label_comparacao.grid(row=14, column=0, columnspan=2)

# Botão para agrupar itens do pedido
tk.Button(frame_interno, text="Agrupar Itens do Pedido", command=agrupar_pedido).grid(row=15, column=0, columnspan=2)

# Campos para calcular aleatório
tk.Label(frame_interno, text="Qtd. Máxima de Produtos:").grid(row=16, column=0)
entry_quantidade_maxima = tk.Entry(frame_interno)
entry_quantidade_maxima.grid(row=16, column=1)

tk.Label(frame_interno, text="Valor Máximo Total:").grid(row=17, column=0)
entry_valor_maximo = tk.Entry(frame_interno)
entry_valor_maximo.grid(row=17, column=1)

# Botão para calcular aleatório
tk.Button(frame_interno, text="Calcular Aleatório", command=calcular_maximo_aleatorio).grid(row=18, column=0, columnspan=2)

# Campo para adicionar um novo produto
tk.Label(frame_interno, text="Código do Produto:").grid(row=19, column=0)
entry_codigo_produto = tk.Entry(frame_interno)
entry_codigo_produto.grid(row=19, column=1)

tk.Label(frame_interno, text="Nome do Produto:").grid(row=20, column=0)
entry_nome_produto = tk.Entry(frame_interno)
entry_nome_produto.grid(row=20, column=1)

tk.Label(frame_interno, text="Valor do Produto (R$):").grid(row=21, column=0)
entry_valor_produto = tk.Entry(frame_interno)
entry_valor_produto.grid(row=21, column=1)

# Botão para adicionar o novo produto
tk.Button(frame_interno, text="Adicionar Produto", command=adicionar_produto).grid(row=22, column=0, columnspan=2)

# Inicia o loop da interface
janela.mainloop()
```
