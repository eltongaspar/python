```python
#Desafio
# Loop infinito que verifica se o usuário gosta de Python
while True:
    resposta = input("Você gosta de Python? ").strip().upper()
    if resposta == "SIM":
        print("Resposta correta")
        break
    else:
        print("Esta não é a resposta correta, tente novamente")
```

```python
#atividade 01


# Atividade 01 usando while

i = 10  # começa em 10
acabou_normalmente = True  # variável para simular o else do for

while i > 0:
    resposta = input(f"{i} (digite 'SAIR' para encerrar): ").strip().upper()

    if resposta == "SAIR":
        print("Você saiu antes do fim!")
        acabou_normalmente = False  # não terminou normalmente
        break

    i -= 1  # decrementa

# Simula o else do for
if acabou_normalmente:
    print("ACABOU")
```

```python
#atvidade 02
# Inicializa a variável acumuladora
soma = 0

while True:
    # Pede um número ao usuário
    numero = int(input("Digite um número (0 para encerrar): "))

    # Verifica se o número digitado é 0
    if numero == 0:
        break  # Sai do loop

    # Soma o número ao acumulado
    soma += numero

# Mostra o resultado final
print(f"O total acumulado da soma é: {soma}")
```

```python
#atividade 03

while True:
    # Pede ao usuário um número ou a palavra SAIR
    entrada = input("Digite um número para ver a tabuada (ou 'SAIR' para encerrar): ").strip().upper()

    # Verifica se o usuário quer sair
    if entrada == "SAIR":
        print("Programa encerrado.")
        break

    # Tenta converter a entrada em número
    try:
        numero = int(entrada)

        # Mostra a tabuada de 1 a 10
        print(f"\nTabuada do {numero}:")
        for i in range(1, 11):
            print(f"{numero} x {i} = {numero * i}")
        print()  # Linha em branco para separar
    except ValueError:
        # Caso o usuário digite algo inválido
        print("Por favor, digite um número válido ou 'SAIR' para encerrar.\n")
```

```python
#atvidade 04

# Senha correta
senha_correta = "1234"

while True:
    # Pede a senha
    senha = input("Digite a senha (ou 'SAIR' para encerrar): ").strip().upper()

    # Verifica se o usuário quer sair
    if senha == "SAIR":
        print("Encerrando programa...")
        break

    # Compara com a senha correta
    if senha == senha_correta:
        print("Acesso permitido!")
        break
    else:
        print("Senha incorreta, tente novamente.")
```

```python
#atividade 05

while True:
    # Pede um número ao usuário
    numero = int(input("Digite um número (0 para encerrar): "))

    # Verifica se o usuário quer encerrar
    if numero == 0:
        print("Programa encerrado.")
        break

    # Verifica se é par ou ímpar
    if numero % 2 == 0:
        print(f"{numero} é PAR.\n")
    else:
        print(f"{numero} é ÍMPAR.\n")
```

```python
# Bonus

# Solicita ao usuário qual número ele quer treinar
numero = int(input("Digite o número que deseja treinar a tabuada: "))

# Inicializa contadores
acertos = 0
erros = 0
multiplicador = 1

while True:
    # Pede a resposta do cálculo ou SAIR
    resposta = input(f"{numero} x {multiplicador} = ? (ou 'SAIR' para encerrar): ").strip().upper()

    # Verifica se o usuário quer sair
    if resposta == "SAIR":
        break

    # Tenta converter a resposta para número
    try:
        resposta = int(resposta)
        resultado_correto = numero * multiplicador

        if resposta == resultado_correto:
            print("CORRETO ✅")
            acertos += 1
        else:
            print(f"QUE PENA, VOCÊ ERROU 😢 O VALOR CORRETO É {resultado_correto}")
            erros += 1

        multiplicador += 1  # passa para o próximo cálculo
    except ValueError:
        print("Digite um número válido ou 'SAIR' para encerrar.")

# Mostra o resultado final
print(f"\nTotal de acertos: {acertos}")
print(f"Total de erros: {erros}")
3333333
```
