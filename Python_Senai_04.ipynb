```python
# atividade 01

# laço com for
for numero in range(1, 101):  # percorre os números de 1 até 99
    resultado = numero % 2    # calcula o resto da divisão por 2
    if resultado == 0:        # se o resto for 0, significa que é par
        print(numero)         # imprime o número par
```

```python
# atividade 02

# variáveis de controle
numero = 0
cont = 0
maior = 0                # começa em 0, será atualizado conforme os números digitados
menor = 99999999999      # começa com um valor bem alto para garantir que o 1º número seja menor
soma = 0                 # acumula a soma dos números digitados
media = 0                # será calculada no final
amostras = 0             # conta quantos números foram digitados

# laço para ler 10 números inteiros
for cont in range(10):
    numero = int(input("Digite um valor inteiro: "))

    # verifica se o número é maior que o atual "maior"
    if numero > maior:
        maior = numero

    # verifica se o número é menor que o atual "menor"
    if numero < menor:
        menor = numero

    # acumula o número na soma total
    soma = soma + numero

    # conta mais uma amostra
    amostras = amostras + 1

# calcula a média após a entrada dos números
media = soma / amostras

# mostra os resultados
print("Maior número: ", maior)
print("Menor número: ", menor)
print("Soma dos números: ", soma)
print("Média dos números: ", media)
```

```python
# atividade 03

# variáveis de controle
temperatura = 0
soma = 0        # acumula a soma das temperaturas
media = 0       # será calculada no final
cont = 0        # conta quantas pessoas tiveram temperatura registrada

# laço infinito, só para quando o usuário decidir
while True:
    temperatura = float(input("Informe a temperatura do paciente: "))

    # classificação da temperatura
    if temperatura <= 37.2:
        print("O paciente está com temperatura normal.")
    elif temperatura >= 37.3 and temperatura <= 38:
        print("O paciente está com febril.")
    elif temperatura >= 38.1 and temperatura <= 39:
        print("O paciente está com febre alta.")
    elif temperatura >= 39.1:
        print("O paciente está com febre muito alta")

    # acumula valores para cálculo posterior
    soma += temperatura
    cont += 1

    # pergunta se quer continuar
    continue_input = input("Você quer inserir outra temperatura? (sim/nao): ").lower()
    if continue_input != 'sim':   # se não for "sim", encerra o laço
        break

# cálculo final da média
media = soma / cont

# resultados
print("A média da temperatura é: ", media)
print("Total de pessoas:", cont)
```

```python
# atividade 04

# variáveis de controle
idade = 0
media_h = 0.0   # média das idades masculinas
media_m = 0.0   # média das idades femininas
soma_h = 0      # soma das idades dos homens
soma_m = 0      # soma das idades das mulheres
cont_m = 0      # contador de homens
cont_f = 0      # contador de mulheres

# laço que repete 10 vezes
for i in range(10):
    sexo = input("Informe sexo M para Masculino e F para Feminino: ").upper()
    idade = int(input("Informe a idade: "))

    if sexo == "M":
        soma_h += idade   # soma idades dos homens
        cont_m += 1       # conta mais um homem
    elif sexo == "F":
        soma_m += idade   # soma idades das mulheres
        cont_f += 1       # conta mais uma mulher
    else:
        print("Sexo inválido. Por favor, digite M ou F.")

# calcular médias, evitando divisão por zero
media_h = soma_h / cont_m if cont_m > 0 else 0
media_m = soma_m / cont_f if cont_f > 0 else 0

# média geral = soma total das idades ÷ total de pessoas
soma_geral = soma_h + soma_m
cont_geral = cont_m + cont_f
media_geral = soma_geral / cont_geral if cont_geral > 0 else 0

# exibir resultados
print("A média de idade dos homens é: ", media_h)
print("A média de idade das mulheres é: ", media_m)
print("A média de idade geral é: ", media_geral)
```

```python
# atividade 5

# variável de controle
cont = 5

# laço que continua enquanto cont for menor que 100
for cont in range(101):
    cont += 1   # incrementa cont em 1 a cada iteração

    # verifica se cont **não é múltiplo de 5** e **é múltiplo de 7**
    if cont % 5 > 0 and cont % 7 == 0:
        print(cont)  # imprime o número que atende às condições
```

```python
# atividade 6
# Programa para inverter uma string usando for

texto = input("Digite uma frase ou palavra: ")

invertido = ""  # variável para guardar o resultado

# percorre o texto de trás para frente
for i in range(len(texto) - 1, -1, -1):
    invertido += texto[i]

print("Texto invertido:", invertido)
```

```python
# Programa de treino de tabuada

# Solicita o número para treino
numero = int(input("Digite o número que deseja treinar a tabuada: "))

# Inicializa contadores
acertos = 0
erros = 0

# Loop de 1 a 10 para a tabuada
for i in range(1, 11):
    resposta = int(input(f"{numero} x {i} = "))
    valor_correto = numero * i
    if resposta == valor_correto:
        print("CORRETO")
        acertos += 1
    else:
        print(f"QUE PENA, VOCÊ ERROU, O VALOR CORRETO É {valor_correto}")
        erros += 1

# Imprime o total de acertos e erros
print(f"Total de acertos: {acertos}")
print(f"Total de erros: {erros}")
```

```python
# atividade 7
# Programa para calcular fatorial

# pede ao usuário um número inteiro
n = int(input("Digite um número para calcular o fatorial: "))

# inicializa a variável fatorial com 1
# (o fatorial é uma multiplicação, e 1 é o elemento neutro da multiplicação)
fatorial = 1

# loop que vai de 1 até n (inclusive)
for i in range(1, n + 1):
    fatorial *= i  # multiplica o valor atual de fatorial por i

# exibe o resultado final
print(f"O fatorial de {n} é {fatorial}")
```
