```python
# Importa a biblioteca 'gradio' para criar interfaces de usuário web interativas.
import gradio as gr

# Obtém a chave da API da Groq armazenada de forma segura nas "User data" do Colab.
# É uma boa prática não embutir chaves de API diretamente no código.
from google.colab import userdata
api_key = userdata.get('GROQ_API_KEY')
```

```python
# Inicializa o cliente Groq usando a chave da API obtida anteriormente.
# Este cliente será usado para fazer chamadas à API da Groq.
client = Groq(api_key=api_key)
```

```python
# Define uma função 'chat_groq' que será usada como o backend do chatbot.
# Ela recebe a mensagem atual do usuário e o histórico da conversa.
def chat_groq(message, history):
    # Formata o histórico da conversa para o formato que a API da Groq espera.
    # Cada par de (usuário, assistente) é adicionado como uma mensagem com seu respectivo 'role'.
    messages = []
    for human, assistant in history:
        messages.append({"role": "user", "content": human})
        messages.append({"role": "assistant", "content": assistant})
    # Adiciona a mensagem atual do usuário à lista de mensagens.
    messages.append({"role": "user", "content": message})

    # Faz a chamada para a API de conclusão de chat da Groq.
    # 'model' especifica qual modelo da Groq será usado (aqui, 'llama-3.3-70b-versatile').
    # 'messages' são as mensagens formatadas da conversa.
    # 'temperature' controla a aleatoriedade da resposta (0.7 é um bom equilíbrio).
    # 'max_tokens' limita o comprimento da resposta gerada.
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile", # Ou "llama3-8b-8192"
        messages=messages,
        temperature=0.7,
        max_tokens=1024,
    )
    # Retorna o conteúdo da primeira (e geralmente única) resposta do assistente.
    return response.choices[0].message.content
```

```python
# Cria e lança uma interface de chat interativa usando Gradio.
# 'fn=chat_groq' especifica que a função 'chat_groq' será chamada para processar as mensagens do usuário.
# 'title' define o título exibido na interface do chat.
# 'description' fornece uma breve descrição do chatbot.
gr.ChatInterface(
    fn=chat_groq,
    title="Groq AI Chatbot",
    description="Chat rápido com Llama 3 da Groq",
).launch()
```
