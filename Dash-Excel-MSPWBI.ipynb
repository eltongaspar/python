```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import clear_output, display
import time
import random

meses = ["Jan","Fev","Mar","Abr","Mai","Jun","Jul","Ago","Set","Out","Nov","Dez"]

titulos_dashboard = [
"Central de Inteligência Empresarial",
"Monitoramento Estratégico de Negócios",
"Análise Corporativa de Dados",
"Painel Executivo de Performance",
"Controle Operacional Analítico"
]

while True:

    receita = np.random.randint(200,800,12)
    custos = np.random.randint(100,500,12)
    lucro = receita - custos
    clientes = np.random.randint(50,200,12)
    transacoes = np.random.randint(80,300,12)
    marketing = np.random.randint(20,150,12)
    vendas_online = np.random.randint(100,400,12)
    vendas_loja = np.random.randint(100,400,12)

    tabela = pd.DataFrame({
        "Mês": meses,
        "Receita": receita,
        "Custos": custos,
        "Lucro": lucro,
        "Clientes": clientes,
        "Transações": transacoes
    })

    graficos = [
        "Receita Mensal",
        "Custos Operacionais",
        "Lucro Mensal",
        "Crescimento de Clientes",
        "Volume de Transações",
        "Investimento em Marketing",
        "Receita vs Custos",
        "Vendas Online",
        "Vendas em Loja",
        "Distribuição Receita x Custos"
    ]

    for grafico in graficos:

        clear_output(wait=True)

        plt.figure(figsize=(12,6))
        plt.title(random.choice(titulos_dashboard) + " - " + grafico, fontsize=16)

        if grafico == "Receita Mensal":
            plt.plot(meses, receita, marker='o', linewidth=3)

        elif grafico == "Custos Operacionais":
            plt.bar(meses, custos)

        elif grafico == "Lucro Mensal":
            plt.plot(meses, lucro, marker='o', linewidth=3)

        elif grafico == "Crescimento de Clientes":
            plt.plot(meses, clientes, marker='o', linewidth=3)
            plt.fill_between(meses, clientes, alpha=0.3)

        elif grafico == "Volume de Transações":
            plt.bar(meses, transacoes)

        elif grafico == "Investimento em Marketing":
            plt.plot(meses, marketing, marker='o')

        elif grafico == "Receita vs Custos":
            plt.plot(meses, receita, label="Receita", linewidth=3)
            plt.plot(meses, custos, label="Custos", linewidth=3)
            plt.legend()

        elif grafico == "Vendas Online":
            plt.bar(meses, vendas_online)

        elif grafico == "Vendas em Loja":
            plt.bar(meses, vendas_loja)

        elif grafico == "Distribuição Receita x Custos":
            plt.pie(
                [sum(receita), sum(custos)],
                labels=["Receita Total","Custos Totais"],
                autopct='%1.1f%%'
            )

        plt.grid(True)
        plt.xlabel("Meses")
        plt.ylabel("Valores")

        plt.show()

        print("\nTabela Operacional (Estilo Excel)\n")
        display(tabela)

        time.sleep(3)
```
