```python
#Desafio
# Programa para separar números pares e ímpares,
# exibindo resultados passo a passo

# Listas para armazenar os números
numeros = []   # todos os números digitados
pares = []     # apenas os números pares
impares = []   # apenas os números ímpares

# Entrada de dados
while True:
    n = int(input("Digite um número (0 para encerrar): "))

    # Se o usuário digitar 0, o loop encerra
    if n == 0:
        break

    # Adiciona o número à lista principal
    numeros.append(n)

    # Já verifica se é par ou ímpar e adiciona na lista correspondente
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

    # Mostra a situação parcial após cada número digitado
    print(f"\nAté agora você digitou: {numeros}")
    print(f"Pares: {pares}")
    print(f"Ímpares: {impares}\n")

# Quando o usuário terminar, ordenamos as listas
numeros.sort()
pares.sort()
impares.sort()

# Impressão dos resultados finais
print("\n===== RESULTADOS FINAIS =====")
print("Lista completa em ordem crescente:", numeros)
print("Números pares:", pares)
print("Números ímpares:", impares)

# Cálculo das somas
print("\nSoma dos números:", sum(numeros))
print("Soma dos pares:", sum(pares))
print("Soma dos ímpares:", sum(impares))
```

```python
#atividae 01

# Programa que lê vários números e calcula soma, multiplicação, maior e menor

# Lista para armazenar os números
numeros = []

# Entrada de dados
while True:
    n = int(input("Digite um número (0 para encerrar): "))

    if n == 0:  # condição de parada
        break

    numeros.append(n)  # adiciona o número na lista

# Verificação: só calcula se a lista não estiver vazia
if numeros:
    # Calcula a soma usando a função sum()
    soma = sum(numeros)

    # Calcula a multiplicação (precisa de um loop)
    multiplicacao = 1
    for num in numeros:
        multiplicacao *= num

    # Maior e menor valores
    maior = max(numeros)
    menor = min(numeros)

    # Exibe os resultados
    print("\n===== RESULTADOS =====")
    print("Números digitados:", numeros)
    print("Soma:", soma)
    print("Multiplicação:", multiplicacao)
    print("Maior número:", maior)
    print("Menor número:", menor)
else:
    print("Nenhum número foi informado!")
```

```python
#atividde02

# Programa que lê várias palavras e conta quantas vezes uma palavra aparece

# Lista para armazenar as palavras digitadas
palavras = []

# Entrada de dados
while True:
    p = input("Digite uma palavra (0 para encerrar): ").strip()

    if p == "0":  # condição de parada
        break

    palavras.append(p)  # adiciona a palavra na lista

# Solicita a palavra que o usuário deseja contar
busca = input("\nDigite a palavra que deseja contar: ").strip()

# Conta quantas vezes a palavra aparece na lista
ocorrencias = palavras.count(busca)

# Exibe os resultados
print("\n===== RESULTADO =====")
print("Lista de palavras digitadas:", palavras)
print(f"A palavra '{busca}' aparece {ocorrencias} vez(es).")
```

```python
#atividade 03

# Programa que lê números e separa em:
# - Lista principal
# - Lista sem repetição
# - Lista com números repetidos

# Lista principal
numeros = []

# Entrada de dados
while True:
    n = int(input("Digite um número (0 para encerrar): "))
    if n == 0:
        break
    numeros.append(n)

# Criar lista sem repetição usando set() (conjunto)
sem_repeticao = list(set(numeros))

# Criar lista de números repetidos
repetidos = []
for num in numeros:
    if numeros.count(num) > 1 and num not in repetidos:
        repetidos.append(num)

# Exibindo os resultados
print("\n===== RESULTADOS =====")
print("Lista completa:", numeros)
print("Números sem repetição:", sem_repeticao)
print("Números repetidos:", repetidos)
```

```python
#atividade 04

 #Programa que solicita vários números ao usuário
# - encerra ao digitar 0
# - cria lista completa, lista sem repetição e lista de repetidos

numeros = []   # lista principal

# Entrada de dados
while True:
    n = int(input("Digite um número (0 para encerrar): "))
    if n == 0:
        break
    numeros.append(n)

# Lista sem repetição
sem_repeticao = list(set(numeros))

# Lista de repetidos
repetidos = []
for num in numeros:
    if numeros.count(num) > 1 and num not in repetidos:
        repetidos.append(num)

# Saída de dados
print("\n--- RESULTADOS ---")
print("Lista completa:", numeros)
print("Sem repetição:", sem_repeticao)
print("Repetidos:", repetidos)
```
