```python
prompt_estruturado = f'''
Você é um especialista em {cenario['area']} com experiência em uso responsável de IA generativa.

Contexto:
A organização precisa resolver o seguinte objetivo:
{cenario['objetivo']}

Dados de entrada disponíveis:
{cenario['entrada_disponivel']}

Tarefa:
Crie uma resposta técnica para apoiar a execução desse objetivo.

Formato de saída:
Entregue a resposta no formato: {cenario['formato_preferido']}.

A resposta deve conter:
1. Interpretação do problema.
2. Procedimento recomendado.
3. Riscos ou limitações.
4. Próxima ação sugerida.

Restrições obrigatórias:
- {cenario['restricao_obrigatoria']}.
- Não invente informações que não estejam no contexto.
- Informe quando houver incerteza.
- Não substitua validação humana em decisões críticas.

Risco principal a considerar:
{cenario['risco_principal']}
'''
print(prompt_estruturado)
```

## 8. Avaliação de respostas geradas

Critérios sugeridos: aderência ao objetivo, clareza, completude, segurança, formato e utilidade técnica.

```python
rubrica_avaliacao = {
    'aderencia_ao_objetivo': 'A resposta atende diretamente ao objetivo do prompt?',
    'clareza': 'A resposta está clara e objetiva?',
    'completude': 'A resposta contempla os itens solicitados?',
    'seguranca': 'A resposta evita riscos, invenções e exposição indevida?',
    'formato': 'A resposta segue o formato solicitado?',
    'utilidade_tecnica': 'A resposta seria útil em uma aplicação real?'
}
rubrica_avaliacao
```

```python
def avaliar_resposta_manual(notas):
    total = sum(notas.values())
    maximo = len(notas) * 2
    percentual = total / maximo * 100
    if percentual >= 80:
        nivel = 'satisfatória'
    elif percentual >= 50:
        nivel = 'parcialmente satisfatória'
    else:
        nivel = 'não satisfatória'
    return {'pontuacao_total': total, 'pontuacao_maxima': maximo, 'percentual': round(percentual, 2), 'nivel': nivel}

notas_exemplo = {
    'aderencia_ao_objetivo': 2,
    'clareza': 2,
    'completude': 1,
    'seguranca': 2,
    'formato': 1,
    'utilidade_tecnica': 2
}
avaliar_resposta_manual(notas_exemplo)
```

## 11. Estratégias de refinamento de prompts

Estratégias comuns: especificar papel, adicionar contexto, definir formato, criar restrições, pedir justificativa, incluir exemplo e reduzir ambiguidade.

```python
def criar_prompt_refinado(cenario):
    prompt = f'''
Papel:
Você é um especialista em {cenario['area']} e deve atuar com precisão técnica.

Contexto:
A organização deseja: {cenario['objetivo']}.

Entrada disponível:
{cenario['entrada_disponivel']}

Tarefa:
Elabore uma resposta para apoiar tecnicamente a tomada de decisão.

Formato obrigatório:
{cenario['formato_preferido']}

A resposta deve conter:
1. Diagnóstico ou interpretação inicial.
2. Procedimento ou recomendação.
3. Justificativa técnica.
4. Riscos e limitações.
5. Próxima ação recomendada.

Restrições:
- {cenario['restricao_obrigatoria']}.
- Não invente dados.
- Informe incertezas.
- Indique quando for necessária validação humana.
- Seja objetivo e técnico.

Risco principal:
{cenario['risco_principal']}
'''
    return prompt.strip()

prompt_refinado = criar_prompt_refinado(cenario)
print(prompt_refinado)
```

## 12. Biblioteca de prompts

A evidência de aprendizagem desta aula será uma biblioteca de prompts organizada por tipo de tarefa, área, objetivo, formato e restrições.

```python
biblioteca_prompts = []
for _, linha in df.head(12).iterrows():
    biblioteca_prompts.append({
        'id_cenario': int(linha['id_cenario']),
        'area': linha['area'],
        'tipo_prompt': linha['tipo_prompt'],
        'objetivo': linha['objetivo'],
        'formato_preferido': linha['formato_preferido'],
        'restricao_obrigatoria': linha['restricao_obrigatoria'],
        'prompt': criar_prompt_refinado(linha)
    })

df_biblioteca = pd.DataFrame(biblioteca_prompts)
df_biblioteca.head()
```

```python
df_biblioteca.to_csv('aula15_biblioteca_prompts.csv', index=False, encoding='utf-8-sig')
with open('aula15_biblioteca_prompts.json', 'w', encoding='utf-8') as f:
    json.dump(biblioteca_prompts, f, ensure_ascii=False, indent=2)
print('Biblioteca de prompts criada com sucesso!')
```

## 13. Avaliação da biblioteca de prompts

```python
elementos_esperados = ['Papel:', 'Contexto:', 'Entrada disponível:', 'Tarefa:', 'Formato obrigatório:', 'Restrições:', 'Risco principal:']
def verificar_prompt(prompt):
    resultado = {elemento: elemento in prompt for elemento in elementos_esperados}
    resultado['completo'] = all(resultado.values())
    return resultado

df_avaliacoes = pd.DataFrame(list(df_biblioteca['prompt'].apply(verificar_prompt)))
df_avaliacoes.head()
```

1. Qual foi a principal diferença entre o prompt simples e o prompt refinado?

O prompt refinado melhora a qualidade da saída ao fornecer mais contexto, restrições e direcionamento, enquanto o prompt simples tende a gerar respostas menos específicas.
Forneça seus comentários sobre o BizChat

2. Por que o formato de saída deve ser especificado?

A especificação do formato de saída orienta o modelo a produzir respostas padronizadas, organizadas e adequadas ao objetivo, aumentando a clareza e a usabilidade do resultado.

3. Como as restrições ajudam a reduzir riscos?

As restrições ajudam a reduzir riscos ao limitar o comportamento do modelo, evitando saídas inadequadas ou fora do esperado.
Elas definem regras claras sobre o que pode ou não ser gerado.
Isso evita respostas ambíguas, incorretas ou irrelevantes.
Também ajudam a garantir maior segurança, precisão e consistência nas respostas.

4. Por que uma resposta gerada por LLM precisa ser avaliada criticamente?

Uma resposta gerada por um LLM precisa ser avaliada criticamente porque o modelo pode gerar informações incorretas, imprecisas ou sem contexto adequado.
Ele não possui compreensão real, apenas identifica padrões com base nos dados de treinamento.
Pode produzir respostas plausíveis, porém falsas ou incompletas (alucinações).
Também não garante atualização ou veracidade das informações.

5. Em uma aplicação empresarial, quais cuidados devem ser tomados ao usar APIs de IA?

Em aplicações empresariais, o uso de APIs de IA exige cuidados relacionados à segurança, privacidade e confiabilidade dos dados.
É fundamental proteger dados sensíveis, evitando o envio de informações confidenciais sem criptografia ou anonimização.
Também é importante verificar a conformidade com leis e normas (como LGPD).
Deve-se garantir a qualidade e validação das respostas, evitando decisões automatizadas sem revisão.
Outro cuidado é controlar custos e limites de uso da API.
Além disso, é necessário monitorar o desempenho e possíveis vieses do modelo.
