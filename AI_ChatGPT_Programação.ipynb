```python
# FORÇA o Colab a usar animação em HTML
import matplotlib
matplotlib.rcParams['animation.html'] = 'jshtml'

import matplotlib.pyplot as plt
import matplotlib.animation as animation
import random
import string
from IPython.display import display

# ---------------- CONFIGURAÇÕES ----------------
COLUNAS = 80
LINHAS = 40
CARACTERES = string.ascii_letters + string.digits + "@#$%&"

TITULO = "IA GENERATIVA"
SUBTITULO = "Aplicada à Programação – ChatGPT"

# ---------------- ESTADO INICIAL ----------------
drops = [random.randint(0, LINHAS) for _ in range(COLUNAS)]

# ---------------- FIGURA ----------------
fig, ax = plt.subplots(figsize=(14, 7))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")
ax.axis("off")
ax.set_xlim(0, COLUNAS)
ax.set_ylim(0, LINHAS)

texts = []

# ---------------- FUNÇÃO DE ANIMAÇÃO ----------------
def update(frame):
    global texts
    for txt in texts:
        txt.remove()
    texts.clear()

    for i in range(COLUNAS):
        char = random.choice(CARACTERES)
        texts.append(
            ax.text(
                i,
                drops[i],
                char,
                color="#00FF00",
                fontsize=12,
                family="monospace"
            )
        )

        drops[i] -= 1
        if drops[i] < 0:
            drops[i] = LINHAS if random.random() > 0.975 else drops[i]

    # TÍTULO
    texts.append(
        ax.text(
            COLUNAS / 2,
            LINHAS - 5,
            TITULO,
            color="cyan",
            fontsize=28,
            ha="center",
            weight="bold"
        )
    )

    # SUBTÍTULO
    texts.append(
        ax.text(
            COLUNAS / 2,
            LINHAS - 8,
            SUBTITULO,
            color="white",
            fontsize=16,
            ha="center"
        )
    )

    return texts

# ---------------- ANIMAÇÃO ----------------
ani = animation.FuncAnimation(
    fig,
    update,
    frames=200,
    interval=80,
    blit=False
)

display(ani)
```
