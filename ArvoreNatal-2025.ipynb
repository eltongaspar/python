```python
import random
import time
from IPython.display import display, HTML, Javascript, clear_output

def cor_aleatoria_hsl():
    h = random.randint(0, 360)
    s = random.randint(60, 100)
    l = random.randint(35, 70)
    return f"hsl({h} {s}% {l}%)"

def gerar_arvore_html(altura=12):
    largura_total = altura * 2 + 1
    linhas_html = []

    # FOLHAS
    for i in range(altura):
        stars = ""
        for _ in range(2 * i + 1):
            stars += f'<span style="color:{cor_aleatoria_hsl()};">*</span>'

        # numero certo de espaços à esquerda
        espacos = " " * (altura - i - 1)
        linhas_html.append(espacos + stars)

    # TRONCO — sempre centralizado
    tronco_texto = "||"
    espacos_tronco = " " * (altura - 1)

    for _ in range(2):
        linhas_html.append(espacos_tronco + f'<span style="color:{cor_aleatoria_hsl()};">{tronco_texto}</span>')

    # gerar bloco centralizado
    conteudo = "\n".join(linhas_html)

    html = f'''
    <div style="width:100%; display:flex; justify-content:center;">
      <pre style="display:inline-block; text-align:left; font-family: monospace;
                  font-size:14px; line-height:1; white-space:pre;">
{conteudo}
      </pre>
    </div>
    '''

    return html


# LOOP DE ANIMAÇÃO
try:
    while True:
        clear_output(wait=True)
        display(HTML(gerar_arvore_html(12)))
        display(Javascript("window.scrollTo(0, document.body.scrollHeight);"))
        time.sleep(0.25)
except KeyboardInterrupt:
    clear_output()
    print("Animação interrompida.")
```

```python
import random
import time
from IPython.display import display, HTML, Javascript, clear_output

# ---------- GERA CORES ----------
def cor_aleatoria_hsl():
    h = random.randint(0, 360)
    s = random.randint(70, 100)
    l = random.randint(50, 70)
    return f"hsl({h} {s}% {l}%)"

# ---------- CRIA UM FOGO DE ARTIFÍCIO ----------
def gerar_fogo_artificio():
    largura = random.randint(4, 12)
    altura = random.randint(3, 6)
    espacos = random.randint(0, 12)

    fogo_html = []
    for i in range(altura):
        linha = "&nbsp;" * espacos
        for _ in range(largura):
            linha += f'<span style="color:{cor_aleatoria_hsl()}; font-weight: bold;">*</span>'
        fogo_html.append(linha)
        espacos = max(0, espacos - 1)

    return "<br>".join(fogo_html)

# ---------- TEXTO COM EFEITO COLORIDO ----------
def texto_colorido(txt):
    html = ""
    for letra in txt:
        cor = cor_aleatoria_hsl()
        html += f'<span style="color:{cor}; font-size:28px; font-weight:bold;">{letra}</span>'
    return html

# ---------- CONTAGEM REGRESSIVA ----------
def contagem_regressiva(segundos=5):
    for i in range(segundos, 0, -1):
        clear_output(wait=True)
        display(HTML(f"""
        <div style="text-align:center; font-size:60px; font-weight:bold; color:{cor_aleatoria_hsl()};">
            {i}
        </div>
        """))
        time.sleep(1)


# ================================
#        ANIMAÇÃO PRINCIPAL
# ================================
try:
    # CONTAGEM
    contagem_regressiva(5)

    # MENSAGEM "FELIZ ANO NOVO!"
    for _ in range(15):
        clear_output(wait=True)
        mensagem = texto_colorido("🎉 FELIZ ANO NOVO! 🎉")
        fogo = gerar_fogo_artificio() + "<br>" + gerar_fogo_artificio()

        display(HTML(f"""
        <div style="text-align:center;">
            <div style="margin-bottom:20px;">{mensagem}</div>
            <pre style="font-family: monospace; font-size: 18px; line-height: 1; white-space: pre-wrap;">
{fogo}
            </pre>
        </div>
        """))
        time.sleep(0.6)

    # LOOP INFINITO DE FOGOS (caso queira continuar)
    while True:
        clear_output(wait=True)
        fogo = gerar_fogo_artificio()
        display(HTML(f'<pre style="font-family: monospace; font-size: 18px; white-space: pre-wrap;">{fogo}</pre>'))
        display(Javascript("window.scrollTo(0, document.body.scrollHeight);"))
        time.sleep(0.5)

except KeyboardInterrupt:
    clear_output()
    print("Animação encerrada.")
```

```python
import random
import time
from IPython.display import clear_output, display, HTML

# -------------------------------------
# Cores HSL vibrantes
# -------------------------------------
def cor():
    return f"hsl({random.randint(0,360)} {random.randint(70,100)}% {random.randint(50,70)}%)"

# -------------------------------------
# Cria uma matriz "tela"
# -------------------------------------
def criar_tela(larg=70, alt=25):
    return [[" " for _ in range(larg)] for _ in range(alt)]

# -------------------------------------
# Converte matriz em HTML colorido
# -------------------------------------
def tela_para_html(tela):
    html = "<pre style='font-family: monospace; font-size:16px; line-height:1;'>"
    for linha in tela:
        for celula in linha:
            if isinstance(celula, tuple):
                char, c = celula
                html += f"<span style='color:{c}; font-weight:bold;'>{char}</span>"
            else:
                html += celula
        html += "\n"
    html += "</pre>"
    return html

# -------------------------------------
# Desenha o foguete subindo
# -------------------------------------
def animar_subida(tela, x, frames=8):
    for i in range(frames):
        clear_output(wait=True)
        tela_temp = criar_tela()

        y = len(tela_temp) - 2 - i
        if 0 <= y < len(tela_temp):
            tela_temp[y][x] = ("|", cor())

        display(HTML(tela_para_html(tela_temp)))
        time.sleep(0.08)

# -------------------------------------
# Explosão circular
# -------------------------------------
def animar_explosao(x, y):
    particulas = []

    for _ in range(35):
        ang = random.uniform(0, 6.28)
        vel = random.uniform(0.5, 1.8)
        particulas.append({
            "x": x,
            "y": y,
            "dx": vel * math.cos(ang),
            "dy": vel * math.sin(ang),
            "cor": cor()
        })

    # anima os fragmentos
    for frame in range(10):
        tela_temp = criar_tela()
        clear_output(wait=True)

        for p in particulas:
            p["x"] += p["dx"]
            p["y"] += p["dy"]
            p["dy"] += 0.1  # gravidade

            xi, yi = int(p["x"]), int(p["y"])
            if 0 <= yi < len(tela_temp) and 0 <= xi < len(tela_temp[0]):
                tela_temp[yi][xi] = ("*", p["cor"])

        display(HTML(tela_para_html(tela_temp)))
        time.sleep(0.09)

# -------------------------------------
# Import math depois (Colab não aceita import antes em bloco)
# -------------------------------------
import math

# -------------------------------------
# ANIMAÇÃO PRINCIPAL
# -------------------------------------
try:
    while True:
        # escolher posição aleatória para o foguete
        x = random.randint(10, 60)

        # anima subida
        animar_subida(None, x, frames=random.randint(5, 9))

        # ponto da explosão
        y_explosao = random.randint(6, 12)

        # anima explosão
        animar_explosao(x, y_explosao)

except KeyboardInterrupt:
    clear_output()
    print("Animação encerrada.")
```
