```python
import random
import time
import math # Necessário para a animação do foguete
from IPython.display import display, HTML, Javascript, clear_output

# --- Funções de Cor (Consolidada) ---
def cor_aleatoria_hsl():
    """Gera uma cor HSL aleatória vibrante."""
    h = random.randint(0, 360)
    s = random.randint(70, 100) # Alta saturação para cores vibrantes
    l = random.randint(50, 70)  # Luminosidade média
    return f"hsl({h} {s}% {l}%)"


# --- Animação 1: Árvore de Natal (Originalmente cell_id: 121223c6) ---
def gerar_arvore_html(altura=12):
    """Gera o HTML para uma árvore de natal colorida."""
    largura_total = altura * 2 + 1
    linhas_html = []

    # FOLHAS
    for i in range(altura):
        stars = ""
        for _ in range(2 * i + 1):
            stars += f'<span style="color:{cor_aleatoria_hsl()};">*</span>'

        # numero certo de espaços à esquerda
        espacos = " " * (altura - i - 1)
        linhas_html.append(espacos + stars)

    # TRONCO — sempre centralizado
    tronco_texto = "||"
    espacos_tronco = " " * (altura - 1)

    for _ in range(2):
        linhas_html.append(espacos_tronco + f'<span style="color:{cor_aleatoria_hsl()};">{tronco_texto}</span>')

    # gerar bloco centralizado
    conteudo = "\n".join(linhas_html)

    html = f'''
    <div style="width:100%; display:flex; justify-content:center;">
      <pre style="display:inline-block; text-align:left; font-family: monospace;
                  font-size:14px; line-height:1; white-space:pre;">
{conteudo}
      </pre>
    </div>
    '''
    return html

def run_christmas_tree_animation(duration_seconds=15):
    """Executa a animação da árvore de natal por um tempo limitado."""
    print("\n--- Iniciando Animação da Árvore de Natal ---")
    start_time = time.time()
    try:
        while time.time() - start_time < duration_seconds:
            clear_output(wait=True)
            display(HTML(gerar_arvore_html(12)))
            display(Javascript("window.scrollTo(0, document.body.scrollHeight);"))
            time.sleep(0.25)
    except KeyboardInterrupt:
        print("Animação da Árvore de Natal interrompida.")
    finally:
        clear_output()
        print("--- Animação da Árvore de Natal Finalizada ---")


# --- Animação 2: Fogos de Artifício de Ano Novo (Originalmente cell_id: BUHpZ2Zspnwv) ---
def gerar_fogo_artificio():
    """Cria uma string HTML que representa um fogo de artifício."""
    largura = random.randint(4, 12)
    altura = random.randint(3, 6)
    espacos = random.randint(0, 12)

    fogo_html = []
    for i in range(altura):
        linha = "&nbsp;" * espacos
        for _ in range(largura):
            linha += f'<span style="color:{cor_aleatoria_hsl()}; font-weight: bold;">*</span>'
        fogo_html.append(linha)
        espacos = max(0, espacos - 1)

    return "<br>".join(fogo_html)

def texto_colorido(txt):
    """Transforma um texto em HTML com cada letra colorida aleatoriamente."""
    html = ""
    for letra in txt:
        cor = cor_aleatoria_hsl()
        html += f'<span style="color:{cor}; font-size:28px; font-weight:bold;">{letra}</span>'
    return html

def contagem_regressiva(segundos=5):
    """Exibe uma contagem regressiva."""
    for i in range(segundos, 0, -1):
        clear_output(wait=True)
        display(HTML(f"""
        <div style="text-align:center; font-size:60px; font-weight:bold; color:{cor_aleatoria_hsl()};">
            {i}
        </div>
        """))
        time.sleep(1)

def run_new_year_fireworks_animation(message_duration_seconds=9, fireworks_duration_seconds=15):
    """Executa a animação de fogos de artifício de ano novo por um tempo limitado."""
    print("\n--- Iniciando Animação de Fogos de Artifício de Ano Novo ---")
    try:
        # CONTAGEM REGRESSIVA
        contagem_regressiva(5)

        # MENSAGEM "FELIZ ANO NOVO!"
        start_message_time = time.time()
        while time.time() - start_message_time < message_duration_seconds:
            clear_output(wait=True)
            mensagem = texto_colorido("🎉 FELIZ ANO NOVO! 🎉")
            fogo = gerar_fogo_artificio() + "<br>" + gerar_fogo_artificio()

            display(HTML(f"""
            <div style="text-align:center;">
                <div style="margin-bottom:20px;">{mensagem}</div>
                <pre style="font-family: monospace; font-size: 18px; line-height: 1; white-space: pre-wrap;">
{fogo}
                </pre>
            </div>
            """))
            time.sleep(0.6)

        # LOOP DE FOGOS (agora limitado)
        start_fireworks_time = time.time()
        while time.time() - start_fireworks_time < fireworks_duration_seconds:
            clear_output(wait=True)
            fogo = gerar_fogo_artificio()
            display(HTML(f'<pre style="font-family: monospace; font-size: 18px; white-space: pre-wrap;">{fogo}</pre>'))
            display(Javascript("window.scrollTo(0, document.body.scrollHeight);"))
            time.sleep(0.5)

    except KeyboardInterrupt:
        print("Animação de Fogos de Artifício de Ano Novo interrompida.")
    finally:
        clear_output()
        print("--- Animação de Fogos de Artifício de Ano Novo Finalizada ---")


# --- Animação 3: Foguetes e Explosões (Originalmente cell_id: tLuSSYcuqxk-) ---
def criar_tela(larg=70, alt=25):
    """Cria uma matriz vazia para a 'tela' da animação."""
    return [[" " for _ in range(larg)] for _ in range(alt)]

def tela_para_html(tela):
    """Converte a matriz da tela em HTML colorido."""
    html = "<pre style='font-family: monospace; font-size:16px; line-height:1;'>"
    for linha in tela:
        for celula in linha:
            if isinstance(celula, tuple):
                char, c = celula
                html += f"<span style='color:{c}; font-weight:bold;'>{char}</span>"
            else:
                html += celula
        html += "\n"
    html += "</pre>"
    return html

def animar_subida(tela_dummy, x, frames=8): # 'tela_dummy' is not used, consistent with original
    """Anima o foguete subindo."""
    for i in range(frames):
        clear_output(wait=True)
        tela_temp = criar_tela()

        y = len(tela_temp) - 2 - i
        if 0 <= y < len(tela_temp):
            tela_temp[y][x] = ("|", cor_aleatoria_hsl()) # Usando cor_aleatoria_hsl

        display(HTML(tela_para_html(tela_temp)))
        time.sleep(0.08)

def animar_explosao(x, y):
    """Anima a explosão circular de um foguete."""
    particulas = []

    for _ in range(35):
        ang = random.uniform(0, 6.28) # 2*PI
        vel = random.uniform(0.5, 1.8)
        particulas.append({
            "x": x,
            "y": y,
            "dx": vel * math.cos(ang),
            "dy": vel * math.sin(ang),
            "cor": cor_aleatoria_hsl() # Usando cor_aleatoria_hsl
        })

    # anima os fragmentos
    for frame in range(10):
        tela_temp = criar_tela()
        clear_output(wait=True)

        for p in particulas:
            p["x"] += p["dx"]
            p["y"] += p["dy"]
            p["dy"] += 0.1  # gravidade

            xi, yi = int(p["x"]), int(p["y"])
            if 0 <= yi < len(tela_temp) and 0 <= xi < len(tela_temp[0]):
                tela_temp[yi][xi] = ("*", p["cor"])

        display(HTML(tela_para_html(tela_temp)))
        time.sleep(0.09)

def run_rocket_fireworks_animation(duration_seconds=20):
    """Executa a animação de foguetes e explosões por um tempo limitado."""
    print("\n--- Iniciando Animação de Foguetes e Explosões ---")
    start_time = time.time()
    try:
        while time.time() - start_time < duration_seconds:
            # escolher posição aleatória para o foguete
            x = random.randint(10, 60)

            # anima subida
            animar_subida(None, x, frames=random.randint(5, 9))

            # ponto da explosão
            y_explosao = random.randint(6, 12)

            # anima explosão
            animar_explosao(x, y_explosao)

    except KeyboardInterrupt:
        print("Animação de Foguetes e Explosões interrompida.")
    finally:
        clear_output()
        print("--- Animação de Foguetes e Explosões Finalizada ---")


# ==================================
#     Execução Sequencial das Animações
# ==================================

# Chamar as animações em sequência
# Cada uma será executada por um tempo definido ou até ser interrompida.
run_christmas_tree_animation(duration_seconds=15) # Animação da árvore por 15 segundos
time.sleep(3) # Pequena pausa entre as animações

run_new_year_fireworks_animation(message_duration_seconds=9, fireworks_duration_seconds=15) # Ano Novo por 9s de mensagem + 15s de fogos
time.sleep(3) # Pequena pausa entre as animações

run_rocket_fireworks_animation(duration_seconds=25) # Foguetes por 25 segundos

print("\nTodas as animações foram executadas. Clique no botão de interrupção (quadrado) para parar a execução a qualquer momento.")
```
