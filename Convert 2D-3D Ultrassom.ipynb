```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.ndimage import gaussian_filter
import os
from PIL import Image

# Caminho para as imagens de ultrassom
image_folder = "E:\Dados\exam_png"  # Substitua pelo caminho correto

# Função para carregar imagens e criar uma matriz volumétrica
def load_ultrasound_images(folder_path):
    images = []
    for filename in sorted(os.listdir(folder_path)):
        if filename.endswith(".png") or filename.endswith(".jpg"):
            img = Image.open(os.path.join(folder_path, filename)).convert("L")
            images.append(np.array(img))
    return np.array(images)

# Carregar as imagens e aplicar suavização
volume = load_ultrasound_images(image_folder)
smoothed_volume = gaussian_filter(volume, sigma=1)  # Suavização para melhorar visualização

# Criar uma representação 3D do volume
def render_3d(volume):
    fig = plt.figure(figsize=(10, 7))
    ax = fig.add_subplot(111, projection='3d')

    # Criar pontos para exibir a densidade do volume
    x, y, z = np.where(volume > 50)  # Threshold para visualizar apenas intensidades relevantes
    ax.scatter(x, y, z, c=volume[x, y, z], cmap='gray', alpha=0.7, s=1)

    ax.set_xlabel("Eixo X")
    ax.set_ylabel("Eixo Y")
    ax.set_zlabel("Eixo Z")
    plt.title("Reconstrução 3D de Imagens de Ultrassom")
    plt.show()

# Renderizar o volume
render_3d(smoothed_volume)
```
