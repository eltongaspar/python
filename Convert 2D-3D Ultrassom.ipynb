```python
import numpy as np
from PIL import Image, ImageEnhance
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter
import os

# Função para carregar e melhorar uma imagem de ultrassom
def enhance_ultrasound(image_path, target_size=(512, 512)):
    # Carregar a imagem e redimensionar
    img = Image.open(image_path).convert("L")
    img_resized = img.resize(target_size)

    # Melhorar o contraste
    enhancer = ImageEnhance.Contrast(img_resized)
    img_contrast = enhancer.enhance(2.0)  # Aumentar o contraste

    # Converter para numpy array
    image_data = np.array(img_contrast)

    # Aplicar um filtro de suavização
    smoothed_image = gaussian_filter(image_data, sigma=1)

    return smoothed_image

# Função para aplicar estilo artístico e exibir a imagem
def artistic_ultrasound(image_data, output_name):
    # Criar a figura
    plt.figure(figsize=(10, 8))

    # Aplicar um mapa de cores artístico
    plt.imshow(image_data, cmap="magma", interpolation="bicubic")

    # Remover os eixos
    plt.axis('off')

    # Título
    plt.title("Ultrassom Estilizado", fontsize=16, color="white")
    plt.savefig(output_name, bbox_inches='tight', pad_inches=0, dpi=300)  # Salvar a imagem estilizada
    plt.show()

# Caminho para o diretório contendo as imagens de ultrassom
image_folder = "E:/Dados/exam_png/"  # Substitua pelo caminho correto

# Iterar sobre todas as imagens no diretório
for filename in os.listdir(image_folder):
    if filename.endswith(".png") or filename.endswith(".jpg"):
        image_path = os.path.join(image_folder, filename)
        print(f"Processando: {filename}")
        enhanced_image = enhance_ultrasound(image_path)
        artistic_ultrasound(enhanced_image, f"stylized_{filename}")


```

```python
import os
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from skimage import exposure

# Função para aplicar CLAHE
def apply_clahe(image_path, target_size=(512, 512)):
    # Carregar a imagem e redimensionar
    img = Image.open(image_path).convert("L")
    img_resized = img.resize(target_size)
    image_data = np.array(img_resized)

    # Aplicar CLAHE
    clahe_image = exposure.equalize_adapthist(image_data / 255.0, clip_limit=0.03) * 255
    return clahe_image

# Função para exibir uma imagem
def show_image(image_data, title="Imagem Processada"):
    plt.figure(figsize=(10, 8))
    plt.imshow(image_data, cmap="gray", interpolation="nearest")
    plt.axis('off')  # Remove os eixos
    plt.title(title, fontsize=16, color="black")
    plt.show()

# Função para processar todas as imagens de um diretório
def process_images_in_directory(input_dir, output_dir, target_size=(512, 512)):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)  # Cria o diretório de saída, se não existir

    for filename in os.listdir(input_dir):
        if filename.endswith(".png") or filename.endswith(".jpg"):
            input_path = os.path.join(input_dir, filename)
            output_path = os.path.join(output_dir, f"processed_{filename}")

            print(f"Processando: {filename}")

            # Processar a imagem
            enhanced_image = apply_clahe(input_path, target_size)

            # Salvar a imagem processada
            enhanced_image_to_save = Image.fromarray(enhanced_image.astype(np.uint8))
            enhanced_image_to_save.save(output_path)
            print(f"Salvo: {output_path}")

            # Visualizar a imagem processada
            show_image(enhanced_image, title=f"Processado: {filename}")

# Caminho para o diretório de entrada e saída
input_directory = "E:/Dados/exam_png/"  # Substitua pelo diretório correto
output_directory = "E:/Dados/exam_png_processed/"  # Diretório para salvar imagens processadas

# Processar todas as imagens do diretório
process_images_in_directory(input_directory, output_directory)
```

```python
import os
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from skimage import feature, exposure

# Função para aplicar CLAHE e/ou detecção de bordas Canny
def process_image(image_path, target_size=(512, 512)):
    # Carregar e redimensionar a imagem
    img = Image.open(image_path).convert("L")
    img_resized = img.resize(target_size)
    image_data = np.array(img_resized)

    # Equalização adaptativa (CLAHE)
    clahe_image = exposure.equalize_adapthist(image_data / 255.0, clip_limit=0.03) * 255

    # Aplicar detecção de bordas (Canny)
    edges = feature.canny(clahe_image.astype(np.uint8), sigma=2)

    # Retornar imagem suavizada (CLAHE) e bordas detectadas
    return clahe_image, edges

# Função para exibir imagens
def show_images(original, processed, edges, filename):
    plt.figure(figsize=(15, 5))

    # Imagem original
    plt.subplot(1, 3, 1)
    plt.imshow(original, cmap="gray")
    plt.title("Original")
    plt.axis("off")

    # Imagem processada (CLAHE)
    plt.subplot(1, 3, 2)
    plt.imshow(processed, cmap="gray")
    plt.title("Processada (CLAHE)")
    plt.axis("off")

    # Bordas detectadas (Canny)
    plt.subplot(1, 3, 3)
    plt.imshow(edges, cmap="gray")
    plt.title("Bordas (Canny)")
    plt.axis("off")

    plt.suptitle(f"Resultados para {filename}", fontsize=16)
    plt.show()

# Caminho para os diretórios de entrada e saída
input_directory = "E:/Dados/exam_png/"  # Substitua pelo caminho do diretório de entrada
output_directory = "E:/Dados/exam_png_processed/"  # Diretório para salvar imagens processadas

# Criar diretório de saída, se não existir
if not os.path.exists(output_directory):
    os.makedirs(output_directory)

# Processar todas as imagens no diretório
for filename in os.listdir(input_directory):
    if filename.endswith(".png") or filename.endswith(".jpg"):
        input_path = os.path.join(input_directory, filename)
        output_path = os.path.join(output_directory, f"processed_{filename}")

        print(f"Processando: {filename}")

        # Processar a imagem
        original_image = Image.open(input_path).convert("L")
        clahe_image, edges = process_image(input_path)

        # Salvar a imagem CLAHE processada
        processed_image_to_save = Image.fromarray(clahe_image.astype(np.uint8))
        processed_image_to_save.save(output_path)

        # Visualizar resultados
        show_images(original_image, clahe_image, edges, filename)

        print(f"Salvo: {output_path}")
```

```python
import os
import numpy as np
from PIL import Image, ImageEnhance
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter
from skimage import exposure, feature

# Função para aplicar CLAHE
def apply_clahe(image_data):
    # Aplicar CLAHE (Equalização Adaptativa)
    clahe_image = exposure.equalize_adapthist(image_data / 255.0, clip_limit=0.03) * 255
    return clahe_image

# Função para aplicar filtro Canny
def apply_canny(image_data):
    # Aplicar detecção de bordas Canny
    edges = feature.canny(image_data.astype(np.uint8), sigma=2)
    return edges

# Função para melhorar imagem com suavização e contraste
def enhance_ultrasound(image_data):
    # Melhorar o contraste
    image_pil = Image.fromarray(image_data.astype(np.uint8))
    enhancer = ImageEnhance.Contrast(image_pil)
    img_contrast = enhancer.enhance(2.0)  # Aumentar o contraste
    # Converter para numpy array e suavizar
    smoothed_image = gaussian_filter(np.array(img_contrast), sigma=1)
    return smoothed_image

# Função para exibir as imagens lado a lado
def show_images(original, clahe, enhanced, edges, filename):
    plt.figure(figsize=(20, 10))

    # Imagem original
    plt.subplot(1, 4, 1)
    plt.imshow(original, cmap="gray")
    plt.title("Original")
    plt.axis("off")

    # Imagem CLAHE
    plt.subplot(1, 4, 2)
    plt.imshow(clahe, cmap="gray")
    plt.title("CLAHE (Equalizado)")
    plt.axis("off")

    # Imagem Melhorada
    plt.subplot(1, 4, 3)
    plt.imshow(enhanced, cmap="magma")
    plt.title("Melhorada (Estilizada)")
    plt.axis("off")

    # Bordas detectadas
    plt.subplot(1, 4, 4)
    plt.imshow(edges, cmap="gray")
    plt.title("Bordas (Canny)")
    plt.axis("off")

    plt.suptitle(f"Processamento de: {filename}", fontsize=16)
    plt.show()

# Função para processar todas as imagens em um diretório
def process_directory(input_dir, output_dir):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)  # Criar diretório de saída, se não existir

    for filename in os.listdir(input_dir):
        if filename.endswith(".png") or filename.endswith(".jpg"):
            input_path = os.path.join(input_dir, filename)
            output_path = os.path.join(output_dir, f"processed_{filename}")

            print(f"Processando: {filename}")

            # Carregar a imagem
            img = Image.open(input_path).convert("L")
            original_image = np.array(img.resize((512, 512)))

            # Aplicar CLAHE
            clahe_image = apply_clahe(original_image)

            # Melhorar e estilizar
            enhanced_image = enhance_ultrasound(original_image)

            # Detecção de bordas
            edges_image = apply_canny(original_image)

            # Salvar a imagem processada CLAHE
            processed_image_to_save = Image.fromarray(clahe_image.astype(np.uint8))
            processed_image_to_save.save(output_path)

            # Exibir as imagens lado a lado
            show_images(original_image, clahe_image, enhanced_image, edges_image, filename)

            print(f"Imagem processada salva em: {output_path}")

# Caminhos para diretórios
input_directory = "E:/Dados/exam_png/"  # Substitua pelo caminho do diretório de entrada
output_directory = "E:/Dados/exam_png_processed/"  # Diretório para salvar imagens processadas

# Processar o diretório
process_directory(input_directory, output_directory)
```

```python
import os
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from skimage import exposure, feature
from scipy.ndimage import sobel, laplace
import cv2

# Função para aplicar CLAHE
def apply_clahe(image_data):
    clahe_image = exposure.equalize_adapthist(image_data / 255.0, clip_limit=0.03) * 255
    return clahe_image

# Função para aplicar Sobel
def apply_sobel(image_data):
    sobel_x = sobel(image_data, axis=0)
    sobel_y = sobel(image_data, axis=1)
    sobel_combined = np.hypot(sobel_x, sobel_y)
    sobel_combined = (sobel_combined / sobel_combined.max()) * 255
    return sobel_combined

# Função para aplicar Laplaciano
def apply_laplacian(image_data):
    laplacian_image = laplace(image_data)
    laplacian_image = (laplacian_image / laplacian_image.max()) * 255
    return laplacian_image

# Função para aplicar Filtro Bilateral
def apply_bilateral_filter(image_data):
    bilateral = cv2.bilateralFilter(image_data.astype(np.uint8), 9, 75, 75)
    return bilateral

# Função para aplicar Correção Gamma
def apply_gamma_correction(image_data, gamma=1.2):
    gamma_corrected = ((image_data / 255.0) ** gamma) * 255
    return gamma_corrected

# Função para aplicar Equalização de Histograma Global
def apply_histogram_equalization(image_data):
    equalized = exposure.equalize_hist(image_data) * 255
    return equalized

# Função para gerar todas as versões da imagem
def process_ultrasound_versions(image_data):
    versions = {
        "Original": image_data,
        "CLAHE": apply_clahe(image_data),
        "Sobel": apply_sobel(image_data),
        "Laplacian": apply_laplacian(image_data),
        "Bilateral": apply_bilateral_filter(image_data),
        "Gamma Correction": apply_gamma_correction(image_data, gamma=1.2),
        "Histogram Equalization": apply_histogram_equalization(image_data),
    }
    return versions

# Função para exibir as imagens lado a lado
def show_all_versions(versions, filename):
    num_versions = len(versions)
    plt.figure(figsize=(20, 5))
    for i, (name, version) in enumerate(versions.items()):
        plt.subplot(1, num_versions, i + 1)
        plt.imshow(version, cmap="gray")
        plt.title(name, fontsize=10)
        plt.axis("off")
    plt.suptitle(f"Comparação de Versões para {filename}", fontsize=16)
    plt.show()

# Função para processar imagens em um diretório
def process_directory(input_dir):
    for filename in os.listdir(input_dir):
        if filename.endswith(".png") or filename.endswith(".jpg"):
            input_path = os.path.join(input_dir, filename)
            print(f"Processando: {filename}")

            # Carregar a imagem
            img = Image.open(input_path).convert("L")
            image_data = np.array(img.resize((512, 512)))

            # Processar versões
            versions = process_ultrasound_versions(image_data)

            # Exibir todas as versões lado a lado
            show_all_versions(versions, filename)

# Caminho para o diretório de entrada
input_directory = "E:/Dados/exam_png/"  # Substitua pelo diretório correto

# Processar o diretório
process_directory(input_directory)
```
