```python
import random

print("--- GERADOR DE NÚMEROS ALEATÓRIOS ---\n")

# Solicita os limites de intervalo apenas uma vez no início
while True:
    try:
        range_inicial = int(input("Defina o limite inferior do intervalo: "))
        range_final = int(input("Defina o limite superior do intervalo: "))

        if range_inicial > range_final:
            print("O limite inferior não pode ser maior que o limite superior. Tente novamente.\n")
            continue
        break
    except ValueError:
        print("Por favor, insira números inteiros válidos para definir o intervalo.\n")

# Solicita os números de exceção apenas uma vez no início
excecoes_input = input("Digite os números de exceção (separados por espaço, ou Enter para nenhum): ")

# Converte as exceções para inteiros
excecoes = set()
for x in excecoes_input.split():
    try:
        excecoes.add(int(x))
    except ValueError:
        pass

# Filtra os números válidos permitidos baseando-se nas exceções iniciais
numeros_possiveis = [num for num in range(range_inicial, range_final + 1) if num not in excecoes]

# Loop para a sequência de sorteios diretos
while True:
    if not numeros_possiveis:
        print("Erro: Todas as opções deste intervalo foram excluídas pelas suas exceções. O programa será encerrado.")
        break

    # Realiza e exibe o sorteio diretamente
    numero_aleatorio = random.choice(numeros_possiveis)
    print(f"\n-> SORTEADO: {numero_aleatorio} (Intervalo: [{range_inicial} até {range_final}] | Excluindo: {list(excecoes) if excecoes else 'Nenhum'})")

    # Pergunta de confirmação de continuidade após exibir o resultado
    opcao = input("Pressione Enter para realizar o próximo sorteio ou digite 'SAIR' para encerrar: ")
    if opcao.upper() == "SAIR":
        print("Programa encerrado com sucesso.")
        break
```
