```python
#Exercicio 4

# declaracao de variaveis
base = float()
altura = float()
area = float()
perimetro = float()

#entrada de dados
base = float(input("Digite o valor da base do retangulo: "))
altura = float(input("Digite o valor da altura do retangulo: "))

#calculos
area = base * altura
perimetro = 2 * (base + altura)

#exibir resultados
print("A area do retangulo é: ", area)
print("O perimetro do retangulo é: ", perimetro)

```

```python
#Exercicio 5

# declaração de variaveis

prestação = float()
valor = float()
tempo = float()
taxa = float()

#entrada de dados
valor = float(input("Digite o valor da conta: "))
taxa = float(input("Digite a taxa de juros: "))
tempo = float(input("Digite o tempo em meses: "))

#calculos
prestação = valor + (valor * (taxa / 100) * tempo)

#exibir resultados
print("O valor da prestaçao é: ", prestação)
```
