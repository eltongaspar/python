```python
import openpyxl
import re

def contains_special_chars(text):
    """Verifica se a string contém caracteres especiais."""
    return bool(re.search(r'[^a-zA-Z0-9 \-.,;:!?()@#]', text))  # Corrigido para não usar escape desnecessário

def check_excel_special_chars(file_path):
    """Abre uma planilha Excel e verifica todas as abas por caracteres especiais, desconsiderando a primeira linha."""
    wb = openpyxl.load_workbook(file_path)
    special_chars_cells = {}
    
    for sheet_name in wb.sheetnames:
        sheet = wb[sheet_name]
        special_chars_cells[sheet_name] = []
        
        for row_idx, row in enumerate(sheet.iter_rows(), 1):
            if row_idx == 1:  # Desconsiderar a primeira linha (índice 1 é a primeira linha)
                continue
            for cell in row:
                if cell.value and isinstance(cell.value, str) and contains_special_chars(cell.value):
                    special_chars_cells[sheet_name].append((cell.coordinate, cell.value))
    
    wb.close()
    return special_chars_cells

def select_file():
    """Solicita ao usuário o caminho do arquivo manualmente."""
    return input("Digite o caminho completo do arquivo Excel: ")

# Exemplo de uso:
file_path = select_file()
if file_path:
    result = check_excel_special_chars(file_path)
    
    has_special_chars = False
    for sheet, cells in result.items():
        if cells:
            has_special_chars = True
            print(f"\nCélulas com caracteres especiais na aba '{sheet}':")
            for coord, value in cells:
                print(f"Célula {coord}: {value}")
    
    if not has_special_chars:
        print("Nenhuma célula com caracteres especiais encontrada em nenhuma aba.")
else:
    print("Nenhum arquivo informado.")
```

```python
import openpyxl
import xlrd
import re
import os
from openpyxl.styles import PatternFill

def contains_special_chars(text):
    """Verifica se a string contém caracteres especiais."""
    return bool(re.search(r'[^a-zA-Z0-9 \-.,;:!?()@#]', text))

def process_xlsx(file_path):
    """Processa arquivo .xlsx: verifica e pinta células com caracteres especiais em vermelho."""
    wb = openpyxl.load_workbook(file_path)
    special_chars_cells = {}
    
    # Define o preenchimento vermelho
    red_fill = PatternFill(start_color="FF0000", end_color="FF0000", fill_type="solid")
    
    for sheet_name in wb.sheetnames:
        sheet = wb[sheet_name]
        special_chars_cells[sheet_name] = []
        
        # Itera ignorando a primeira linha (min_row=2)
        for row in sheet.iter_rows(min_row=2):
            for cell in row:
                if cell.value and isinstance(cell.value, str) and contains_special_chars(cell.value):
                    cell.fill = red_fill
                    special_chars_cells[sheet_name].append((cell.coordinate, cell.value))
    
    wb.save(file_path)
    wb.close()
    return special_chars_cells

def process_xls(file_path):
    """Processa arquivo .xls: lê com xlrd, cria novo arquivo .xlsx com formatação e pinta as células em vermelho."""
    rb = xlrd.open_workbook(file_path)
    special_chars_cells = {}
    
    red_fill = PatternFill(start_color="FF0000", end_color="FF0000", fill_type="solid")
    new_wb = openpyxl.Workbook()
    # Remove a planilha padrão criada automaticamente
    default_sheet = new_wb.active
    new_wb.remove(default_sheet)
    
    for sheet in rb.sheets():
        sheet_name = sheet.name
        special_chars_cells[sheet_name] = []
        new_sheet = new_wb.create_sheet(title=sheet_name)
        
        # Copia os dados da planilha xlrd para openpyxl
        for row_idx in range(sheet.nrows):
            for col_idx in range(sheet.ncols):
                value = sheet.cell_value(row_idx, col_idx)
                # openpyxl usa índices 1-based
                cell = new_sheet.cell(row=row_idx+1, column=col_idx+1, value=value)
                # Se não for a primeira linha e for string com caracteres especiais, pinta em vermelho
                if row_idx >= 1 and isinstance(value, str) and contains_special_chars(value):
                    cell.fill = red_fill
                    special_chars_cells[sheet_name].append((cell.coordinate, value))
    
    new_file_path = os.path.splitext(file_path)[0] + "_converted.xlsx"
    new_wb.save(new_file_path)
    new_wb.close()
    print(f"Arquivo convertido e salvo como: {new_file_path}")
    return special_chars_cells

def check_excel_special_chars(file_path):
    """Verifica todas as abas da planilha por caracteres especiais e pinta as células em vermelho."""
    file_extension = os.path.splitext(file_path)[1].lower()
    if file_extension == ".xls":
        return process_xls(file_path)
    else:
        return process_xlsx(file_path)

def select_file():
    """Solicita ao usuário o caminho do arquivo manualmente."""
    return input("Digite o caminho completo do arquivo Excel: ")

# Exemplo de uso:
file_path = select_file()
if file_path:
    result = check_excel_special_chars(file_path)
    
    has_special_chars = False
    for sheet, cells in result.items():
        if cells:
            has_special_chars = True
            print(f"\nCélulas com caracteres especiais na aba '{sheet}':")
            for coord, value in cells:
                print(f"Célula {coord}: {value}")
    
    if not has_special_chars:
        print("Nenhuma célula com caracteres especiais encontrada em nenhuma aba.")
else:
    print("Nenhum arquivo informado.")
```

```python
import openpyxl
import xlrd
import re
import os
from openpyxl.styles import PatternFill
from openpyxl.comments import Comment

def contains_special_chars(text):
    """Verifica se a string contém caracteres especiais."""
    return bool(re.search(r'[^a-zA-Z0-9 \-.,;:!?()@#]', text))

def process_xlsx(file_path):
    """Processa arquivo .xlsx: verifica, pinta células com caracteres especiais em vermelho e adiciona comentário."""
    wb = openpyxl.load_workbook(file_path)
    special_chars_cells = {}
    
    # Define o preenchimento vermelho
    red_fill = PatternFill(start_color="FF0000", end_color="FF0000", fill_type="solid")
    comment_text = "Contém caracteres especiais, isso causa problemas na carga de dados"
    
    for sheet_name in wb.sheetnames:
        sheet = wb[sheet_name]
        special_chars_cells[sheet_name] = []
        
        # Itera ignorando a primeira linha (min_row=2)
        for row in sheet.iter_rows(min_row=2):
            for cell in row:
                if cell.value and isinstance(cell.value, str) and contains_special_chars(cell.value):
                    cell.fill = red_fill
                    cell.comment = Comment(comment_text, "Sistema")
                    special_chars_cells[sheet_name].append((cell.coordinate, cell.value))
    
    wb.save(file_path)
    wb.close()
    return special_chars_cells

def process_xls(file_path):
    """Processa arquivo .xls: lê com xlrd, cria novo arquivo .xlsx com formatação, pinta células em vermelho e adiciona comentário."""
    rb = xlrd.open_workbook(file_path)
    special_chars_cells = {}
    
    red_fill = PatternFill(start_color="FF0000", end_color="FF0000", fill_type="solid")
    comment_text = "Contém caracteres especiais, isso causa problemas na carga de dados"
    new_wb = openpyxl.Workbook()
    
    # Remove a planilha padrão criada automaticamente
    default_sheet = new_wb.active
    new_wb.remove(default_sheet)
    
    for sheet in rb.sheets():
        sheet_name = sheet.name
        special_chars_cells[sheet_name] = []
        new_sheet = new_wb.create_sheet(title=sheet_name)
        
        # Copia os dados da planilha xlrd para openpyxl
        for row_idx in range(sheet.nrows):
            for col_idx in range(sheet.ncols):
                value = sheet.cell_value(row_idx, col_idx)
                cell = new_sheet.cell(row=row_idx+1, column=col_idx+1, value=value)
                
                # Se não for a primeira linha e for string com caracteres especiais, pinta em vermelho e adiciona comentário
                if row_idx >= 1 and isinstance(value, str) and contains_special_chars(value):
                    cell.fill = red_fill
                    cell.comment = Comment(comment_text, "Sistema")
                    special_chars_cells[sheet_name].append((cell.coordinate, value))
    
    new_file_path = os.path.splitext(file_path)[0] + "_converted.xlsx"
    new_wb.save(new_file_path)
    new_wb.close()
    print(f"Arquivo convertido e salvo como: {new_file_path}")
    return special_chars_cells

def check_excel_special_chars(file_path):
    """Verifica todas as abas da planilha por caracteres especiais, pinta células em vermelho e adiciona comentário."""
    file_extension = os.path.splitext(file_path)[1].lower()
    if file_extension == ".xls":
        return process_xls(file_path)
    else:
        return process_xlsx(file_path)

def select_file():
    """Solicita ao usuário o caminho do arquivo manualmente."""
    return input("Digite o caminho completo do arquivo Excel: ")

# Exemplo de uso:
file_path = select_file()
if file_path:
    result = check_excel_special_chars(file_path)
    
    has_special_chars = False
    for sheet, cells in result.items():
        if cells:
            has_special_chars = True
            print(f"\nCélulas com caracteres especiais na aba '{sheet}':")
            for coord, value in cells:
                print(f"Célula {coord}: {value}")
    
    if not has_special_chars:
        print("Nenhuma célula com caracteres especiais encontrada em nenhuma aba.")
else:
    print("Nenhum arquivo informado.")
```
