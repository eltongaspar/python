```python
# Programa para remover números pares de uma lista

# Exemplo de lista L
L = [10, 15, 22, 33, 42, 55, 60, 71]

# Criando uma nova lista apenas com os ímpares
L_impares = [x for x in L if x % 2 != 0]

print("Lista original:", L)
print("Lista sem pares:", L_impares)
```

```python
# Programa para comparar médias das provas P1 e P2

# Exemplo de listas de notas
P1 = [7.5, 8.0, 6.5, 9.0, 5.5]
P2 = [8.5, 7.0, 6.0, 9.5, 6.5]

# Calculando as médias
media_P1 = sum(P1) / len(P1)
media_P2 = sum(P2) / len(P2)

print(f"Média da Prova 1: {media_P1:.2f}")
print(f"Média da Prova 2: {media_P2:.2f}")

# Comparando as médias
if media_P1 > media_P2:
    print("A turma obteve melhor média na Prova 1.")
elif media_P2 > media_P1:
    print("A turma obteve melhor média na Prova 2.")
else:
    print("As médias das duas provas foram iguais.")
```

```python
# Programa para concatenar listas e ordenar

# Exemplo de listas
L1 = [5, 2, 9, 1]
L2 = [8, 3, 7, 4, 6]

# Concatenação
L3 = L1 + L2

print("Lista concatenada:", L3)

# Ordenada de forma crescente
L3_crescente = sorted(L3)
print("Lista em ordem crescente:", L3_crescente)

# Ordenada de forma decrescente
L3_decrescente = sorted(L3, reverse=True)
print("Lista em ordem decrescente:", L3_decrescente)
```

```python
# Criando uma tupla
t = ("maçã", "banana", "laranja", "uva")

print("Tupla original:", t)

# Convertendo para lista
lista = list(t)

# Alterando o último elemento
lista[-1] = 3

# Convertendo de volta para tupla
t = tuple(lista)

print("Tupla modificada:", t)
```
